```python
# ============================================
# CELL 4 - DATASET STRUCTURE INSPECTION
# ============================================

import pandas as pd
import os

# Training files
train_files = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv"
]

# Testing files
test_files = [
    "test_source1.tsv",
    "test_source2.tsv",
    "test_source3.tsv"
]


# Function to inspect TSV file
def inspect_tsv(folder, filename):

    file_path = os.path.join(folder, filename)

    print("\n" + "=" * 70)
    print("FILE:", filename)
    print("=" * 70)

    # Read only first 5 rows
    df = pd.read_csv(
        file_path,
        sep="\t",
        nrows=5
    )

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 Rows:")
    display(df)

    print("\nData Types:")
    print(df.dtypes)

    # File size
    size_mb = os.path.getsize(file_path) / (1024 * 1024)

    print(f"\nFile Size: {size_mb:.2f} MB")


# ============================================
# TRAIN DATA
# ============================================

print("\n\n########## TRAIN DATA ##########")

for file in train_files:
    inspect_tsv(TRAIN_PATH, file)


# ============================================
# TEST DATA
# ============================================

print("\n\n########## TEST DATA ##########")

for file in test_files:
    inspect_tsv(TEST_PATH, file)
```

```python
# ============================================
# CELL 5 - CHUNK-WISE DATA PROCESSING
# ============================================

import pandas as pd
import os

CHUNK_SIZE = 100_000


def process_file_in_chunks(file_path, chunk_size=CHUNK_SIZE):

    total_rows = 0
    chunk_number = 0

    print("\nProcessing:", os.path.basename(file_path))

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=chunk_size
    ):
        chunk_number += 1
        rows = len(chunk)
        total_rows += rows

        print(
            f"Chunk {chunk_number}: "
            f"{rows:,} rows | "
            f"Total processed: {total_rows:,}"
        )

    print(
        f"Completed: {os.path.basename(file_path)} | "
        f"Total rows: {total_rows:,}"
    )

    return total_rows


# ============================================
# TRAIN DATA
# ============================================

print("========== TRAIN DATA ==========")

train_row_counts = {}

for filename in train_files:

    file_path = os.path.join(TRAIN_PATH, filename)

    train_row_counts[filename] = process_file_in_chunks(
        file_path
    )


# ============================================
# TEST DATA
# ============================================

print("\n========== TEST DATA ==========")

test_row_counts = {}

for filename in test_files:

    file_path = os.path.join(TEST_PATH, filename)

    test_row_counts[filename] = process_file_in_chunks(
        file_path
    )


# ============================================
# SUMMARY
# ============================================

print("\n\n========== ROW COUNT SUMMARY ==========")

print("\nTRAIN FILES:")

for filename, count in train_row_counts.items():
    print(f"{filename}: {count:,} rows")

print("\nTEST FILES:")

for filename, count in test_row_counts.items():
    print(f"{filename}: {count:,} rows")
```

```python
# ============================================
# CELL 7 - DATA QUALITY INSPECTION
# ============================================

import pandas as pd
import os

CHUNK_SIZE = 100_000

source_files = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv"
]

def inspect_data_quality(folder, filename):

    file_path = os.path.join(folder, filename)

    print("\n" + "=" * 70)
    print("DATA QUALITY:", filename)
    print("=" * 70)

    total_rows = 0
    missing_counts = {}
    empty_counts = {}
    duplicate_rows = 0

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):
        total_rows += len(chunk)

        # Missing values
        for column in chunk.columns:
            missing_counts[column] = (
                missing_counts.get(column, 0)
                + chunk[column].isna().sum()
            )

            # Empty / whitespace-only values
            empty_counts[column] = (
                empty_counts.get(column, 0)
                + chunk[column].fillna("").str.strip().eq("").sum()
            )

        # Duplicate rows
        duplicate_rows += chunk.duplicated().sum()

    print(f"\nTotal Rows: {total_rows:,}")

    print("\nMissing Values:")
    for column, count in missing_counts.items():
        print(f"{column}: {count:,}")

    print("\nEmpty / Blank Values:")
    for column, count in empty_counts.items():
        print(f"{column}: {count:,}")

    print(f"\nDuplicate Rows: {duplicate_rows:,}")


# ============================================
# RUN INSPECTION
# ============================================

for filename in source_files:
    inspect_data_quality(
        TRAIN_PATH,
        filename
    )
```

```python
# ============================================
# CELL 8 - CLEANING + TEXT NORMALIZATION
# ============================================

import pandas as pd
import os
import re
import unicodedata

CHUNK_SIZE = 100_000

# Create processed data folder
PROCESSED_PATH = os.path.join(
    BASE_PATH,
    "processed_data"
)

os.makedirs(PROCESSED_PATH, exist_ok=True)


# --------------------------------------------
# TEXT NORMALIZATION FUNCTION
# --------------------------------------------

def normalize_text(value):

    if pd.isna(value):
        return ""

    value = str(value)

    # Unicode normalization
    value = unicodedata.normalize("NFKC", value)

    # Convert to lowercase
    value = value.lower()

    # Remove leading/trailing spaces
    value = value.strip()

    # Replace multiple spaces with one space
    value = re.sub(r"\s+", " ", value)

    return value


# --------------------------------------------
# PROCESS ONE TSV FILE
# --------------------------------------------

def clean_and_normalize_file(input_path, output_path):

    first_chunk = True
    total_rows = 0

    print("\nProcessing:", os.path.basename(input_path))

    for chunk in pd.read_csv(
        input_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):

        # ------------------------------------
        # Handle missing values
        # ------------------------------------

        for column in [
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ]:

            if column in chunk.columns:
                chunk[column] = chunk[column].fillna("")


        # ------------------------------------
        # Normalize text columns
        # ------------------------------------

        for column in [
            "business_name",
            "business_address",
            "country"
        ]:

            if column in chunk.columns:
                chunk[column] = chunk[column].apply(
                    normalize_text
                )


        # ------------------------------------
        # Remove exact duplicate rows
        # ------------------------------------

        chunk = chunk.drop_duplicates()


        # ------------------------------------
        # Store processed chunk
        # ------------------------------------

        chunk.to_csv(
            output_path,
            sep="\t",
            index=False,
            mode="w" if first_chunk else "a",
            header=first_chunk
        )

        first_chunk = False

        total_rows += len(chunk)

        print(
            f"Processed: {total_rows:,} rows"
        )

        del chunk

    print(
        "Completed:",
        os.path.basename(output_path)
    )


# ============================================
# PROCESS TRAIN SOURCES
# ============================================

source_files = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv"
]

for filename in source_files:

    input_file = os.path.join(
        TRAIN_PATH,
        filename
    )

    output_file = os.path.join(
        PROCESSED_PATH,
        "cleaned_" + filename
    )

    clean_and_normalize_file(
        input_file,
        output_file
    )


print("\n================================")
print("CLEANING + NORMALIZATION DONE")
print("================================")
```

```python
# ============================================
# CELL 10 - BLOCKING KEY ANALYSIS
# ============================================

import pandas as pd
import os
import re

CHUNK_SIZE = 100_000

source_files = [
    "cleaned_train_source1.tsv",
    "cleaned_train_source2.tsv",
    "cleaned_train_source3.tsv"
]


def analyze_blocking_keys(file_path):

    print("\n" + "=" * 70)
    print("BLOCKING ANALYSIS:", os.path.basename(file_path))
    print("=" * 70)

    total_rows = 0

    country_counts = {}
    name_prefix_counts = {}
    name_token_counts = {}

    address_available = 0
    name_available = 0

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):

        total_rows += len(chunk)

        # ------------------------------------
        # Country analysis
        # ------------------------------------

        countries = (
            chunk["country"]
            .fillna("")
            .str.strip()
        )

        for country, count in countries.value_counts().items():

            if country != "":
                country_counts[country] = (
                    country_counts.get(country, 0) + count
                )

        # ------------------------------------
        # Business name availability
        # ------------------------------------

        names = (
            chunk["business_name"]
            .fillna("")
            .str.strip()
        )

        name_available += (names != "").sum()

        # ------------------------------------
        # Address availability
        # ------------------------------------

        addresses = (
            chunk["business_address"]
            .fillna("")
            .str.strip()
        )

        address_available += (addresses != "").sum()

        # ------------------------------------
        # Name prefix
        # ------------------------------------

        prefixes = (
            names
            .str.replace(r"[^a-z0-9]", "", regex=True)
            .str[:3]
        )

        for prefix, count in prefixes.value_counts().items():

            if prefix != "":
                name_prefix_counts[prefix] = (
                    name_prefix_counts.get(prefix, 0) + count
                )

        # ------------------------------------
        # First token of business name
        # ------------------------------------

        first_tokens = (
            names
            .str.split()
            .str[0]
            .fillna("")
        )

        for token, count in first_tokens.value_counts().items():

            if token != "":
                name_token_counts[token] = (
                    name_token_counts.get(token, 0) + count
                )

    # ========================================
    # RESULTS
    # ========================================

    print(f"\nTotal Rows: {total_rows:,}")

    print(
        f"\nBusiness Name Available: "
        f"{name_available:,} "
        f"({name_available / total_rows * 100:.2f}%)"
    )

    print(
        f"Business Address Available: "
        f"{address_available:,} "
        f"({address_available / total_rows * 100:.2f}%)"
    )

    # ----------------------------------------
    # Country distribution
    # ----------------------------------------

    country_series = (
        pd.Series(country_counts)
        .sort_values(ascending=False)
    )

    print("\nTop 20 Countries:")

    display(
        country_series.head(20).to_frame(
            "count"
        )
    )

    # ----------------------------------------
    # Name prefix distribution
    # ----------------------------------------

    prefix_series = (
        pd.Series(name_prefix_counts)
        .sort_values(ascending=False)
    )

    print("\nTop 20 Name Prefixes:")

    display(
        prefix_series.head(20).to_frame(
            "count"
        )
    )

    # ----------------------------------------
    # First token distribution
    # ----------------------------------------

    token_series = (
        pd.Series(name_token_counts)
        .sort_values(ascending=False)
    )

    print("\nTop 20 Business Name First Tokens:")

    display(
        token_series.head(20).to_frame(
            "count"
        )
    )


# ============================================
# RUN FOR ALL 3 SOURCES
# ============================================

for filename in source_files:

    file_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    analyze_blocking_keys(file_path)
```

```python
# ============================================
# CELL 11 - MEANINGFUL NAME TOKEN ANALYSIS
# ============================================

import pandas as pd
import os
import re
from collections import Counter

CHUNK_SIZE = 100_000

# Common / generic business-name tokens
GENERIC_TOKENS = {
    "the", "private", "pvt", "llc", "inc", "incorporated",
    "limited", "ltd", "corp", "corporation",
    "company", "co", "services", "group",
    "mr", "mrs", "ms", "smt", "dr",
    "shri", "sri", "m/s", "m",
    "and", "of", "for"
}


def extract_meaningful_tokens(text):

    if pd.isna(text):
        return []

    text = str(text).lower()

    # Keep letters and numbers
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    tokens = text.split()

    # Remove generic tokens and very short tokens
    tokens = [
        token
        for token in tokens
        if token not in GENERIC_TOKENS
        and len(token) >= 3
    ]

    return tokens


def analyze_meaningful_tokens(file_path):

    print("\n" + "=" * 70)
    print("MEANINGFUL TOKEN ANALYSIS:", os.path.basename(file_path))
    print("=" * 70)

    token_counter = Counter()
    total_rows = 0
    rows_with_token = 0

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):

        names = chunk["business_name"].fillna("")

        for name in names:

            tokens = extract_meaningful_tokens(name)

            if tokens:
                rows_with_token += 1
                token_counter.update(set(tokens))

        total_rows += len(chunk)

    print(f"\nTotal Rows: {total_rows:,}")

    print(
        f"Rows with meaningful token: "
        f"{rows_with_token:,}"
    )

    print("\nTop 30 Meaningful Tokens:")

    token_df = pd.DataFrame(
        token_counter.most_common(30),
        columns=["token", "count"]
    )

    display(token_df)


# ============================================
# RUN FOR ALL 3 SOURCES
# ============================================

for filename in [
    "cleaned_train_source1.tsv",
    "cleaned_train_source2.tsv",
    "cleaned_train_source3.tsv"
]:

    file_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    analyze_meaningful_tokens(file_path)
```

```python
# ============================================
# CELL 12 - BUILD BLOCKING INDEX
# ============================================

import pandas as pd
import os
import re
from collections import Counter

CHUNK_SIZE = 100_000

# --------------------------------------------
# Blocking configuration
# --------------------------------------------

MIN_TOKEN_LENGTH = 4

# Very common / generic tokens
GENERIC_TOKENS = {
    "the", "private", "pvt", "llc", "inc",
    "incorporated", "limited", "ltd",
    "corp", "corporation", "company", "co",
    "services", "service", "group",
    "mr", "mrs", "ms", "smt", "dr",
    "shri", "sri", "m/s", "and", "of", "for"
}


# --------------------------------------------
# Extract tokens
# --------------------------------------------

def get_tokens(text):

    if pd.isna(text):
        return []

    text = str(text).lower()

    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    tokens = text.split()

    tokens = [
        token
        for token in tokens
        if len(token) >= MIN_TOKEN_LENGTH
        and token not in GENERIC_TOKENS
    ]

    return list(set(tokens))


# ============================================
# STEP 1
# Count token frequency in Source 2 + Source 3
# ============================================

print("=" * 70)
print("STEP 1 - TOKEN FREQUENCY")
print("=" * 70)

token_frequency = Counter()

source_files = [
    "cleaned_train_source2.tsv",
    "cleaned_train_source3.tsv"
]

for filename in source_files:

    file_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    print("\nScanning:", filename)

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):

        names = chunk["business_name"].fillna("")

        for name in names:

            tokens = get_tokens(name)

            token_frequency.update(tokens)

        del chunk

print("\nTotal unique tokens:")
print(f"{len(token_frequency):,}")


# ============================================
# STEP 2
# Keep only reasonably selective tokens
# ============================================

MAX_TOKEN_FREQUENCY = 50_000

useful_tokens = {
    token
    for token, count in token_frequency.items()
    if count <= MAX_TOKEN_FREQUENCY
}

print("\nUseful blocking tokens:")
print(f"{len(useful_tokens):,}")

print(
    f"\nRemoved frequent tokens: "
    f"{len(token_frequency) - len(useful_tokens):,}"
)


# ============================================
# STEP 3
# Save useful tokens
# ============================================

BLOCKING_PATH = os.path.join(
    BASE_PATH,
    "blocking"
)

os.makedirs(
    BLOCKING_PATH,
    exist_ok=True
)

useful_tokens_file = os.path.join(
    BLOCKING_PATH,
    "useful_tokens.txt"
)

with open(
    useful_tokens_file,
    "w",
    encoding="utf-8"
) as f:

    for token in sorted(useful_tokens):
        f.write(token + "\n")


print(
    "\nUseful token list saved to:"
)
print(useful_tokens_file)
```

```python
# ============================================
# CELL 13 - BUILD COUNTRY + TOKEN INDEX
# ============================================

import pandas as pd
import os
import re

CHUNK_SIZE = 100_000

# Load useful tokens
with open(
    useful_tokens_file,
    "r",
    encoding="utf-8"
) as f:
    useful_tokens = set(
        line.strip()
        for line in f
        if line.strip()
    )

print(
    f"Loaded useful tokens: "
    f"{len(useful_tokens):,}"
)


def get_tokens(text):

    if pd.isna(text):
        return []

    text = str(text).lower()

    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    tokens = text.split()

    tokens = [
        token
        for token in set(tokens)
        if len(token) >= 4
        and token in useful_tokens
    ]

    return tokens


# --------------------------------------------
# Output directory
# --------------------------------------------

INDEX_PATH = os.path.join(
    BLOCKING_PATH,
    "source_index"
)

os.makedirs(
    INDEX_PATH,
    exist_ok=True
)


# --------------------------------------------
# Process Source 2 and Source 3
# --------------------------------------------

for filename in [
    "cleaned_train_source2.tsv",
    "cleaned_train_source3.tsv"
]:

    input_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    source_name = (
        "source2"
        if "source2" in filename
        else "source3"
    )

    output_path = os.path.join(
        INDEX_PATH,
        f"{source_name}_blocking_index.tsv"
    )

    print("\n" + "=" * 70)
    print("BUILDING INDEX:", source_name)
    print("=" * 70)

    first_write = True
    total_index_rows = 0

    for chunk in pd.read_csv(
        input_path,
        sep="\t",
        chunksize=CHUNK_SIZE,
        dtype=str
    ):

        records = []

        for _, row in chunk.iterrows():

            country = str(
                row["country"]
            ).strip()

            entity_id = str(
                row["entity_id"]
            ).strip()

            tokens = get_tokens(
                row["business_name"]
            )

            for token in tokens:

                records.append({
                    "block_key": f"{country}|{token}",
                    "entity_id": entity_id
                })

        if records:

            index_chunk = pd.DataFrame(records)

            index_chunk = (
                index_chunk
                .drop_duplicates()
            )

            index_chunk.to_csv(
                output_path,
                sep="\t",
                index=False,
                mode="w" if first_write else "a",
                header=first_write
            )

            first_write = False

            total_index_rows += len(
                index_chunk
            )

        del chunk

    print(
        f"Index rows created: "
        f"{total_index_rows:,}"
    )

    print(
        "Saved:",
        output_path
    )

print("\n========================================")
print("BLOCKING INDEX BUILD COMPLETE")
print("========================================")
```

```python
# ============================================
# CELL 14 - RAM-SAFE CANDIDATE TEST
# ============================================

import pandas as pd
import os
import re

BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

PROCESSED_PATH = os.path.join(
    BASE_PATH,
    "processed_data"
)

BLOCKING_PATH = os.path.join(
    BASE_PATH,
    "blocking"
)

SOURCE1_PATH = os.path.join(
    PROCESSED_PATH,
    "cleaned_train_source1.tsv"
)

SOURCE2_INDEX_PATH = os.path.join(
    BLOCKING_PATH,
    "source_index",
    "source2_blocking_index.tsv"
)

SOURCE3_INDEX_PATH = os.path.join(
    BLOCKING_PATH,
    "source_index",
    "source3_blocking_index.tsv"
)

USEFUL_TOKENS_PATH = os.path.join(
    BLOCKING_PATH,
    "useful_tokens.txt"
)

SAMPLE_ROWS = 100_000
INDEX_CHUNK_SIZE = 100_000


# ============================================
# Check files
# ============================================

print("=" * 70)
print("FILE CHECK")
print("=" * 70)

files_to_check = {
    "Source 1": SOURCE1_PATH,
    "Source 2 index": SOURCE2_INDEX_PATH,
    "Source 3 index": SOURCE3_INDEX_PATH,
    "Useful tokens": USEFUL_TOKENS_PATH
}

for name, path in files_to_check.items():

    print(f"\n{name}:")
    print(path)

    if os.path.exists(path):
        print("STATUS: OK")
    else:
        print("STATUS: NOT FOUND")
        raise FileNotFoundError(path)


# ============================================
# Load useful tokens
# ============================================

with open(
    USEFUL_TOKENS_PATH,
    "r",
    encoding="utf-8"
) as f:

    useful_tokens = set(
        line.strip()
        for line in f
        if line.strip()
    )

print(
    f"\nUseful tokens loaded: "
    f"{len(useful_tokens):,}"
)


# ============================================
# Token function
# ============================================

def get_tokens(text):

    if pd.isna(text):
        return []

    text = str(text).lower()

    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    tokens = text.split()

    return list(set(
        token
        for token in tokens
        if len(token) >= 4
        and token in useful_tokens
    ))


# ============================================
# Load ONLY 100K Source-1 rows
# ============================================

print("\n" + "=" * 70)
print("LOADING SOURCE-1 SAMPLE")
print("=" * 70)

source1_sample = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=SAMPLE_ROWS,
    dtype=str
)

print(
    f"Source-1 sample rows: "
    f"{len(source1_sample):,}"
)


# ============================================
# Create blocking keys
# ============================================

print("\nCreating blocking keys...")

required_keys = set()

for row in source1_sample.itertuples(
    index=False
):

    country = str(
        row.country
    ).strip().lower()

    tokens = get_tokens(
        row.business_name
    )

    for token in tokens:

        required_keys.add(
            f"{country}|{token}"
        )

print(
    f"Unique blocking keys required: "
    f"{len(required_keys):,}"
)


# ============================================
# Candidate output folder
# ============================================

CANDIDATE_DIR = os.path.join(
    BLOCKING_PATH,
    "candidates"
)

os.makedirs(
    CANDIDATE_DIR,
    exist_ok=True
)

CANDIDATE_PATH = os.path.join(
    CANDIDATE_DIR,
    "candidate_test_100k.tsv"
)

if os.path.exists(CANDIDATE_PATH):
    os.remove(CANDIDATE_PATH)


# ============================================
# Scan index chunk-by-chunk
# ============================================

def scan_index(
    index_path,
    source_name
):

    total_rows = 0
    matched_rows = 0
    first_write = True

    print("\n" + "=" * 70)
    print(
        f"SCANNING {source_name.upper()} INDEX"
    )
    print("=" * 70)

    for chunk in pd.read_csv(
        index_path,
        sep="\t",
        chunksize=INDEX_CHUNK_SIZE,
        dtype=str
    ):

        total_rows += len(chunk)

        matched = chunk[
            chunk["block_key"].isin(
                required_keys
            )
        ]

        if not matched.empty:

            matched = matched.copy()

            matched["candidate_source"] = (
                source_name
            )

            matched.to_csv(
                CANDIDATE_PATH,
                sep="\t",
                index=False,
                mode="w" if first_write else "a",
                header=first_write
            )

            first_write = False

            matched_rows += len(matched)

        del matched
        del chunk

        if total_rows % 1_000_000 == 0:

            print(
                f"Scanned: "
                f"{total_rows:,} | "
                f"Matched: "
                f"{matched_rows:,}"
            )

    print(
        f"{source_name} completed."
    )

    print(
        f"Total scanned: "
        f"{total_rows:,}"
    )

    print(
        f"Total matched: "
        f"{matched_rows:,}"
    )

    return matched_rows


# ============================================
# Source 2
# ============================================

source2_matches = scan_index(
    SOURCE2_INDEX_PATH,
    "source2"
)


# ============================================
# Source 3
# ============================================

source3_matches = scan_index(
    SOURCE3_INDEX_PATH,
    "source3"
)


# ============================================
# Final result
# ============================================

total_candidates = (
    source2_matches +
    source3_matches
)

print("\n" + "=" * 70)
print("CANDIDATE TEST COMPLETE")
print("=" * 70)

print(
    f"\nSource-1 records tested: "
    f"{len(source1_sample):,}"
)

print(
    f"Source-2 candidates: "
    f"{source2_matches:,}"
)

print(
    f"Source-3 candidates: "
    f"{source3_matches:,}"
)

print(
    f"Total candidate rows: "
    f"{total_candidates:,}"
)

print(
    f"Average candidates per Source-1: "
    f"{total_candidates / len(source1_sample):.2f}"
)

print("\nCandidate file:")
print(CANDIDATE_PATH)
```

```python
# ============================================
# CELL 16 - BUILD DISK-SAFE BLOCK LOOKUP
# ============================================

import pandas as pd
import os
import sqlite3

BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

BLOCKING_PATH = os.path.join(
    BASE_PATH,
    "blocking"
)

INDEX_PATH = os.path.join(
    BLOCKING_PATH,
    "source_index"
)

DB_PATH = os.path.join(
    BLOCKING_PATH,
    "blocking_lookup.db"
)

SOURCE2_INDEX_PATH = os.path.join(
    INDEX_PATH,
    "source2_blocking_index.tsv"
)

SOURCE3_INDEX_PATH = os.path.join(
    INDEX_PATH,
    "source3_blocking_index.tsv"
)

# Remove old database if it exists
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("BUILDING DISK-BASED BLOCK LOOKUP")
print("=" * 70)


def build_table(
    index_path,
    table_name
):

    print(
        f"\nBuilding table: {table_name}"
    )

    first_chunk = True
    total_rows = 0

    for chunk in pd.read_csv(
        index_path,
        sep="\t",
        chunksize=100_000,
        dtype=str
    ):

        chunk = chunk[
            ["block_key", "entity_id"]
        ]

        chunk = chunk.drop_duplicates()

        chunk.to_sql(
            table_name,
            conn,
            if_exists=(
                "replace"
                if first_chunk
                else "append"
            ),
            index=False
        )

        first_chunk = False
        total_rows += len(chunk)

        if total_rows % 1_000_000 == 0:
            print(
                f"Loaded: {total_rows:,}"
            )

        del chunk

    print(
        f"Completed: {total_rows:,} rows"
    )


# Source 2
build_table(
    SOURCE2_INDEX_PATH,
    "source2_blocks"
)

# Source 3
build_table(
    SOURCE3_INDEX_PATH,
    "source3_blocks"
)


# --------------------------------------------
# Create indexes
# --------------------------------------------

print("\nCreating database indexes...")

conn.execute("""
CREATE INDEX idx_source2_block
ON source2_blocks(block_key)
""")

conn.execute("""
CREATE INDEX idx_source3_block
ON source3_blocks(block_key)
""")

conn.commit()

conn.close()

print("\n" + "=" * 70)
print("DISK-BASED BLOCK LOOKUP COMPLETE")
print("=" * 70)

print("\nDatabase:")
print(DB_PATH)

print(
    f"\nDatabase size: "
    f"{os.path.getsize(DB_PATH) / (1024**3):.2f} GB"
)
```

```python
# ============================================
# CELL 21 - ANALYZE BLOCK SIZE THRESHOLDS
# ============================================

import sqlite3
import pandas as pd

DB_PATH = "/content/blocking_lookup.db"

conn = sqlite3.connect(DB_PATH, timeout=120)

print("=" * 70)
print("BLOCK SIZE THRESHOLD ANALYSIS")
print("=" * 70)

thresholds = [5, 10, 20, 50, 100, 250, 500, 1000]

for table, source in [
    ("source2_block_counts", "SOURCE-2"),
    ("source3_block_counts", "SOURCE-3")
]:

    print("\n" + "=" * 70)
    print(source)
    print("=" * 70)

    rows = []

    for threshold in thresholds:

        result = conn.execute(
            f"""
            SELECT
                COUNT(*) AS blocks,
                COALESCE(SUM(candidate_count), 0) AS rows
            FROM {table}
            WHERE candidate_count <= ?
            """,
            (threshold,)
        ).fetchone()

        blocks, candidate_rows = result

        rows.append({
            "max_block_size": threshold,
            "usable_blocks": blocks,
            "candidate_rows": candidate_rows
        })

    df = pd.DataFrame(rows)

    display(df)

conn.close()

print("\n" + "=" * 70)
print("CELL 21 COMPLETE")
print("=" * 70)
```

```python
# ============================================
# CELL 22 - SAFE CANDIDATE GENERATION TEST
# 10,000 SOURCE-1 RECORDS ONLY
# ============================================

import sqlite3
import pandas as pd
import os
import re
import time

DB_PATH = "/content/blocking_lookup.db"

SOURCE1_PATH = os.path.join(
    PROCESSED_PATH,
    "cleaned_train_source1.tsv"
)

OUTPUT_DIR = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

OUTPUT_PATH = os.path.join(
    OUTPUT_DIR,
    "candidate_test_10k.tsv"
)

MAX_BLOCK_SIZE = 20
TEST_ROWS = 10_000

print("=" * 70)
print("SAFE CANDIDATE GENERATION TEST")
print("=" * 70)

print(f"\nSource-1 test rows : {TEST_ROWS:,}")
print(f"Maximum block size : {MAX_BLOCK_SIZE}")

# ------------------------------------------------
# Load Source-1 sample
# ------------------------------------------------

source1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=TEST_ROWS,
    dtype=str
).fillna("")

print(
    f"\nLoaded Source-1 rows: "
    f"{len(source1):,}"
)

# ------------------------------------------------
# Create blocking keys
# ------------------------------------------------

def make_block_keys(name, address, country):

    text = f"{name} {address}".lower()

    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )

    tokens = set(
        token
        for token in text.split()
        if len(token) >= 3
    )

    keys = []

    country = str(country).strip().lower()

    for token in tokens:
        if country:
            keys.append(
                f"{country}|{token}"
            )

    return keys


# ------------------------------------------------
# Open database
# ------------------------------------------------

conn = sqlite3.connect(
    DB_PATH,
    timeout=120
)

print("\nSQLite database connected.")

# ------------------------------------------------
# Candidate generation
# ------------------------------------------------

candidate_rows = []

start_time = time.time()

for idx, row in source1.iterrows():

    keys = make_block_keys(
        row["business_name"],
        row["business_address"],
        row["country"]
    )

    # Remove duplicate keys
    keys = list(set(keys))

    for block_key in keys:

        # -------------------------------
        # Source-2
        # -------------------------------

        s2 = conn.execute(
            """
            SELECT entity_id
            FROM source2_blocks
            WHERE block_key = ?
            AND block_key IN (
                SELECT block_key
                FROM source2_block_counts
                WHERE block_key = ?
                AND candidate_count <= ?
            )
            """,
            (
                block_key,
                block_key,
                MAX_BLOCK_SIZE
            )
        ).fetchall()

        for (entity_id,) in s2:

            candidate_rows.append({
                "source1_entity_id":
                    row["entity_id"],
                "candidate_entity_id":
                    entity_id,
                "candidate_source":
                    "source2",
                "block_key":
                    block_key
            })

        # -------------------------------
        # Source-3
        # -------------------------------

        s3 = conn.execute(
            """
            SELECT entity_id
            FROM source3_blocks
            WHERE block_key = ?
            AND block_key IN (
                SELECT block_key
                FROM source3_block_counts
                WHERE block_key = ?
                AND candidate_count <= ?
            )
            """,
            (
                block_key,
                block_key,
                MAX_BLOCK_SIZE
            )
        ).fetchall()

        for (entity_id,) in s3:

            candidate_rows.append({
                "source1_entity_id":
                    row["entity_id"],
                "candidate_entity_id":
                    entity_id,
                "candidate_source":
                    "source3",
                "block_key":
                    block_key
            })

    # Progress
    if (idx + 1) % 1000 == 0:

        print(
            f"Processed: {idx + 1:,} | "
            f"Candidates: {len(candidate_rows):,}"
        )

conn.close()

# ------------------------------------------------
# Save candidates
# ------------------------------------------------

candidate_df = pd.DataFrame(
    candidate_rows
)

candidate_df.to_csv(
    OUTPUT_PATH,
    sep="\t",
    index=False
)

elapsed = time.time() - start_time

print("\n" + "=" * 70)
print("CANDIDATE TEST COMPLETE")
print("=" * 70)

print(
    f"\nSource-1 records: "
    f"{TEST_ROWS:,}"
)

print(
    f"Candidate rows: "
    f"{len(candidate_df):,}"
)

print(
    f"Average candidates per Source-1: "
    f"{len(candidate_df) / TEST_ROWS:.2f}"
)

print(
    f"\nTime taken: "
    f"{elapsed / 60:.2f} minutes"
)

print(
    f"\nCandidate file:\n"
    f"{OUTPUT_PATH}"
)
```

```python
# ============================================
# CELL 24 - BLOCKING RECALL VALIDATION
# ============================================

import pandas as pd
import os

BASE_PATH = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

CANDIDATE_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "candidate_test_10k.tsv"
)

GROUND_TRUTH_PATH = os.path.join(
    BASE_PATH,
    "dataset",
    "train",
    "train_ground_truth.tsv"
)

print("=" * 70)
print("BLOCKING RECALL VALIDATION")
print("=" * 70)

# ============================================================
# STEP 1 - LOAD CANDIDATES
# ============================================================

print("\nLoading candidate file...")

candidates = pd.read_csv(
    CANDIDATE_PATH,
    sep="\t",
    dtype=str
)

print(f"Candidate rows: {len(candidates):,}")

# Remove duplicate pairs
candidates = candidates.drop_duplicates(
    subset=[
        "source1_entity_id",
        "candidate_entity_id"
    ]
)

print(
    f"Unique candidate pairs: "
    f"{len(candidates):,}"
)

# ============================================================
# STEP 2 - LOAD GROUND TRUTH
# ============================================================

print("\nLoading ground truth...")

ground_truth = pd.read_csv(
    GROUND_TRUTH_PATH,
    sep="\t",
    nrows=10_000,
    dtype=str
)

print(
    f"Ground-truth rows: "
    f"{len(ground_truth):,}"
)

print("\nGround-truth columns:")
print(ground_truth.columns.tolist())

# ============================================================
# STEP 3 - CREATE CANDIDATE LOOKUP
# ============================================================

print("\nCreating candidate lookup...")

candidate_pairs = set(
    zip(
        candidates["source1_entity_id"],
        candidates["candidate_entity_id"]
    )
)

print(
    f"Candidate lookup pairs: "
    f"{len(candidate_pairs):,}"
)

# ============================================================
# STEP 4 - VALIDATE GROUND-TRUTH MATCHES
# ============================================================

print("\nChecking ground-truth matches...")

total_positive_pairs = 0
captured_positive_pairs = 0

records_with_match = 0
records_with_captured_match = 0

for _, row in ground_truth.iterrows():

    source1_id = str(
        row["source1_entity_id"]
    ).strip()

    matched_ids = row["matched_entity_ids"]

    # --------------------------------------------
    # No match
    # --------------------------------------------

    if pd.isna(matched_ids):
        continue

    matched_ids = str(
        matched_ids
    ).strip()

    if matched_ids == "":
        continue

    # --------------------------------------------
    # Convert matched IDs into list
    # --------------------------------------------

    matched_list = [
        x.strip()
        for x in matched_ids.split(",")
        if x.strip()
    ]

    if len(matched_list) == 0:
        continue

    records_with_match += 1

    record_captured = False

    # --------------------------------------------
    # Check every true match
    # --------------------------------------------

    for target_id in matched_list:

        total_positive_pairs += 1

        pair = (
            source1_id,
            target_id
        )

        if pair in candidate_pairs:

            captured_positive_pairs += 1
            record_captured = True

    if record_captured:

        records_with_captured_match += 1


# ============================================================
# STEP 5 - CALCULATE RECALL
# ============================================================

if total_positive_pairs > 0:

    pair_recall = (
        captured_positive_pairs /
        total_positive_pairs
    ) * 100

else:

    pair_recall = 0.0


if records_with_match > 0:

    record_recall = (
        records_with_captured_match /
        records_with_match
    ) * 100

else:

    record_recall = 0.0


# ============================================================
# STEP 6 - RESULTS
# ============================================================

print("\n" + "=" * 70)
print("BLOCKING RECALL RESULTS")
print("=" * 70)

print(
    f"\nGround-truth positive pairs : "
    f"{total_positive_pairs:,}"
)

print(
    f"Captured positive pairs     : "
    f"{captured_positive_pairs:,}"
)

print(
    f"Missed positive pairs       : "
    f"{total_positive_pairs - captured_positive_pairs:,}"
)

print(
    f"\nPair-level blocking recall  : "
    f"{pair_recall:.2f}%"
)

print(
    f"\nSource-1 records with match : "
    f"{records_with_match:,}"
)

print(
    f"Records with >=1 captured   : "
    f"{records_with_captured_match:,}"
)

print(
    f"Records with no candidate   : "
    f"{records_with_match - records_with_captured_match:,}"
)

print(
    f"\nRecord-level blocking recall: "
    f"{record_recall:.2f}%"
)

print("\n" + "=" * 70)
print("CELL 24 COMPLETE")
print("=" * 70)
```

```python
import os
import sqlite3
import pandas as pd
import re

print("=" * 70)
print("CELL 33 - MISSED MATCH DIAGNOSIS")
print("=" * 70)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------
BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

SOURCE1_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source1.tsv"
)

SOURCE2_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source2.tsv"
)

SOURCE3_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source3.tsv"
)

GT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "ground_truth_aligned_10k.tsv"
)

CANDIDATE_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "candidate_test_10k_v2.tsv"
)

DB_PATH = "/content/blocking_lookup.db"

# ------------------------------------------------------------
# LOAD SOURCE-1 TEST DATA
# ------------------------------------------------------------
print("\nLoading Source-1 test data...")

source1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=10_000,
    dtype=str,
    keep_default_na=False
)

source1_map = source1.set_index("entity_id").to_dict("index")

print("Source-1 records:", f"{len(source1):,}")

# ------------------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------------------
print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
)

# ------------------------------------------------------------
# LOAD CANDIDATES
# ------------------------------------------------------------
print("Loading candidates...")

candidates = pd.read_csv(
    CANDIDATE_PATH,
    sep="\t",
    dtype=str
)

candidate_pairs = set(
    zip(
        candidates["source1_entity_id"],
        candidates["candidate_entity_id"]
    )
)

# ------------------------------------------------------------
# FIND MISSED TRUE PAIRS
# ------------------------------------------------------------
print("\nFinding missed true matches...")

missed_pairs = []

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.isna(matched):
        continue

    matched = str(matched).strip()

    if not matched:
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if not target_id:
            continue

        pair = (s1_id, target_id)

        if pair not in candidate_pairs:
            missed_pairs.append(pair)

print(
    "Missed true pairs:",
    f"{len(missed_pairs):,}"
)

# ------------------------------------------------------------
# TAKE SMALL SAMPLE FOR DIAGNOSIS
# ------------------------------------------------------------
sample_pairs = missed_pairs[:100]

print(
    "Diagnosing first:",
    len(sample_pairs),
    "missed pairs"
)

# ------------------------------------------------------------
# SQLITE CONNECTION
# ------------------------------------------------------------
print("\nConnecting SQLite...")

conn = sqlite3.connect(
    DB_PATH,
    timeout=60
)

conn.execute("PRAGMA busy_timeout = 60000")

# ------------------------------------------------------------
# TOKEN FUNCTION
# ------------------------------------------------------------
def tokenize(text):

    if text is None:
        return set()

    text = str(text).lower()

    return set(
        re.findall(
            r"[a-z0-9]+",
            text
        )
    )

# ------------------------------------------------------------
# DIAGNOSE
# ------------------------------------------------------------
results = []

for s1_id, target_id in sample_pairs:

    # --------------------------------------------------------
    # SOURCE-1 DATA
    # --------------------------------------------------------
    s1 = source1_map.get(s1_id)

    if s1 is None:
        results.append({
            "source1_id": s1_id,
            "target_id": target_id,
            "target_source": (
                "source2"
                if target_id.startswith("S2-")
                else "source3"
            ),
            "status": "SOURCE1_NOT_FOUND",
            "common_tokens": "",
            "common_count": 0,
            "block_sizes": ""
        })
        continue

    # --------------------------------------------------------
    # TARGET SOURCE
    # --------------------------------------------------------
    if target_id.startswith("S2-"):
        table = "source2_blocks"
        freq_table = "source2_block_frequency"
        target_source = "source2"

    elif target_id.startswith("S3-"):
        table = "source3_blocks"
        freq_table = "source3_block_frequency"
        target_source = "source3"

    else:
        results.append({
            "source1_id": s1_id,
            "target_id": target_id,
            "target_source": "unknown",
            "status": "INVALID_TARGET_ID",
            "common_tokens": "",
            "common_count": 0,
            "block_sizes": ""
        })
        continue

    # --------------------------------------------------------
    # GET TARGET BLOCK KEYS
    # --------------------------------------------------------
    target_rows = conn.execute(
        f"""
        SELECT block_key
        FROM {table}
        WHERE entity_id = ?
        """,
        (target_id,)
    ).fetchall()

    target_keys = set(
        row[0]
        for row in target_rows
    )

    # --------------------------------------------------------
    # SOURCE-1 RAW TOKENS
    # --------------------------------------------------------
    s1_tokens = (
        tokenize(s1["business_name"])
        |
        tokenize(s1["business_address"])
    )

    country = str(
        s1["country"]
    ).strip().lower()

    # Expected blocking keys from Source-1
    s1_keys = set()

    for token in s1_tokens:
        s1_keys.add(
            f"{country}|{token}"
        )

    # --------------------------------------------------------
    # COMMON BLOCK KEYS
    # --------------------------------------------------------
    common_keys = s1_keys.intersection(
        target_keys
    )

    # --------------------------------------------------------
    # BLOCK SIZES
    # --------------------------------------------------------
    block_sizes = []

    for key in common_keys:

        row = conn.execute(
            f"""
            SELECT block_size
            FROM {freq_table}
            WHERE block_key = ?
            """,
            (key,)
        ).fetchone()

        if row:
            block_sizes.append(
                f"{key}:{row[0]}"
            )

    # --------------------------------------------------------
    # STATUS
    # --------------------------------------------------------
    if len(common_keys) == 0:

        status = "NO_COMMON_BLOCK"

    elif any(
        int(x.split(":")[-1]) <= 20
        for x in block_sizes
    ):

        status = "SHOULD_BE_CAPTURED"

    else:

        status = "BLOCK_TOO_LARGE"

    results.append({
        "source1_id": s1_id,
        "target_id": target_id,
        "target_source": target_source,
        "status": status,
        "common_tokens": "|".join(
            sorted(common_keys)
        ),
        "common_count": len(common_keys),
        "block_sizes": "|".join(
            block_sizes
        )
    })

conn.close()

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------
diagnosis = pd.DataFrame(results)

print("\n" + "=" * 70)
print("DIAGNOSIS SUMMARY")
print("=" * 70)

print(
    diagnosis["status"].value_counts(
        dropna=False
    )
)

print("\n" + "=" * 70)
print("SAMPLE DIAGNOSIS")
print("=" * 70)

print(
    diagnosis.head(30).to_string(
        index=False
    )
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------
OUTPUT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "missed_match_diagnosis_100.tsv"
)

diagnosis.to_csv(
    OUTPUT_PATH,
    sep="\t",
    index=False
)

print("\nDiagnosis saved to:")
print(OUTPUT_PATH)

print("\n" + "=" * 70)
print("CELL 33 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# CELL 34 - CONTROLLED BLOCK SIZE THRESHOLD EXPERIMENT
# ============================================================

import os
import sqlite3
import pandas as pd
import re
import time

print("=" * 70)
print("CELL 34 - BLOCK SIZE THRESHOLD EXPERIMENT")
print("=" * 70)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------
BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

SOURCE1_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source1.tsv"
)

GT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "ground_truth_aligned_10k.tsv"
)

DB_PATH = "/content/blocking_lookup.db"

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------
TEST_ROWS = 10_000

THRESHOLDS = [
    20,
    100,
    500,
    1000
]

# ------------------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------------------
print("\nLoading Source-1...")

source1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=TEST_ROWS,
    dtype=str,
    keep_default_na=False
)

print(
    "Source-1 records:",
    f"{len(source1):,}"
)

# ------------------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------------------
print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
)

true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.isna(matched):
        continue

    matched = str(matched).strip()

    if not matched:
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add(
                (s1_id, target_id)
            )

print(
    "Ground-truth positive pairs:",
    f"{len(true_pairs):,}"
)

# ------------------------------------------------------------
# LOAD SQLITE
# ------------------------------------------------------------
print("\nConnecting SQLite...")

conn = sqlite3.connect(
    DB_PATH,
    timeout=60
)

conn.execute(
    "PRAGMA busy_timeout = 60000"
)

# ------------------------------------------------------------
# TOKEN FUNCTION
# ------------------------------------------------------------
def get_tokens(text):

    if not text:
        return set()

    return set(
        re.findall(
            r"[a-z0-9]+",
            str(text).lower()
        )
    )

# ------------------------------------------------------------
# SOURCE-1 BLOCK KEYS
# ------------------------------------------------------------
print("\nCreating Source-1 blocking keys...")

source1_blocks = {}

for _, row in source1.iterrows():

    s1_id = row["entity_id"]

    country = str(
        row["country"]
    ).strip().lower()

    tokens = (
        get_tokens(row["business_name"])
        |
        get_tokens(row["business_address"])
    )

    keys = set()

    for token in tokens:

        if country:
            keys.add(
                f"{country}|{token}"
            )

    source1_blocks[s1_id] = keys

print(
    "Source-1 blocking keys created:",
    f"{len(source1_blocks):,}"
)

# ------------------------------------------------------------
# EXPERIMENT
# ------------------------------------------------------------
results = []

for threshold in THRESHOLDS:

    print("\n" + "=" * 70)
    print(
        f"TESTING MAX_BLOCK_SIZE = {threshold}"
    )
    print("=" * 70)

    start = time.time()

    candidate_pairs = set()

    processed = 0

    for s1_id, block_keys in source1_blocks.items():

        # ----------------------------------------------------
        # SOURCE 2
        # ----------------------------------------------------
        for block_key in block_keys:

            row = conn.execute(
                """
                SELECT block_size
                FROM source2_block_frequency
                WHERE block_key = ?
                """,
                (block_key,)
            ).fetchone()

            if row is None:
                continue

            if row[0] > threshold:
                continue

            rows = conn.execute(
                """
                SELECT entity_id
                FROM source2_blocks
                WHERE block_key = ?
                """,
                (block_key,)
            )

            for (entity_id,) in rows:

                candidate_pairs.add(
                    (s1_id, entity_id)
                )

        # ----------------------------------------------------
        # SOURCE 3
        # ----------------------------------------------------
        for block_key in block_keys:

            row = conn.execute(
                """
                SELECT block_size
                FROM source3_block_frequency
                WHERE block_key = ?
                """,
                (block_key,)
            ).fetchone()

            if row is None:
                continue

            if row[0] > threshold:
                continue

            rows = conn.execute(
                """
                SELECT entity_id
                FROM source3_blocks
                WHERE block_key = ?
                """,
                (block_key,)
            )

            for (entity_id,) in rows:

                candidate_pairs.add(
                    (s1_id, entity_id)
                )

        processed += 1

        if processed % 2000 == 0:

            print(
                f"Processed: {processed:,} | "
                f"Candidates: {len(candidate_pairs):,}"
            )

    # --------------------------------------------------------
    # RECALL
    # --------------------------------------------------------
    captured = true_pairs.intersection(
        candidate_pairs
    )

    pair_recall = (
        len(captured)
        / len(true_pairs)
        * 100
    )

    true_source1_ids = {
        s1_id
        for s1_id, _ in true_pairs
    }

    captured_source1_ids = {
        s1_id
        for s1_id, _ in captured
    }

    record_recall = (
        len(
            captured_source1_ids.intersection(
                true_source1_ids
            )
        )
        / len(true_source1_ids)
        * 100
    )

    elapsed = time.time() - start

    avg_candidates = (
        len(candidate_pairs)
        / TEST_ROWS
    )

    # --------------------------------------------------------
    # SAVE RESULT
    # --------------------------------------------------------
    results.append({
        "max_block_size": threshold,
        "candidate_pairs": len(candidate_pairs),
        "avg_candidates_per_record": round(
            avg_candidates, 2
        ),
        "true_pairs": len(true_pairs),
        "captured_pairs": len(captured),
        "pair_recall_percent": round(
            pair_recall, 2
        ),
        "record_recall_percent": round(
            record_recall, 2
        ),
        "time_minutes": round(
            elapsed / 60,
            2
        )
    })

    print("\nRESULT:")
    print(
        "Candidate pairs:",
        f"{len(candidate_pairs):,}"
    )
    print(
        "Average candidates/record:",
        f"{avg_candidates:.2f}"
    )
    print(
        "Captured true pairs:",
        f"{len(captured):,}"
    )
    print(
        "Pair recall:",
        f"{pair_recall:.2f}%"
    )
    print(
        "Record recall:",
        f"{record_recall:.2f}%"
    )
    print(
        "Time:",
        f"{elapsed/60:.2f} minutes"
    )

# ------------------------------------------------------------
# CLOSE DATABASE
# ------------------------------------------------------------
conn.close()

# ------------------------------------------------------------
# FINAL COMPARISON TABLE
# ------------------------------------------------------------
results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("THRESHOLD EXPERIMENT RESULTS")
print("=" * 70)

print(
    results_df.to_string(
        index=False
    )
)

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------
OUTPUT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "threshold_experiment_10k.tsv"
)

results_df.to_csv(
    OUTPUT_PATH,
    sep="\t",
    index=False
)

print("\nResults saved to:")
print(OUTPUT_PATH)

print("\n" + "=" * 70)
print("CELL 34 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CELL 35 - ALTERNATIVE BLOCKING KEY EXPERIMENT
# ================================================================

import os
import re
import sqlite3
import time
import pandas as pd

print("=" * 70)
print("CELL 35 - ALTERNATIVE BLOCKING KEY EXPERIMENT")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------
BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

SOURCE1_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source1.tsv"
)

GT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "ground_truth_aligned_10k.tsv"
)

DB_PATH = "/content/blocking_lookup.db"

TEST_ROWS = 10_000

# IMPORTANT:
# We do NOT use 500/1000 here.
MAX_BLOCK_SIZE = 500


# ================================================================
# LOAD SOURCE-1
# ================================================================

print("\nLoading Source-1...")

source1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=TEST_ROWS,
    dtype=str,
    keep_default_na=False
)

print(
    "Source-1 records:",
    f"{len(source1):,}"
)


# ================================================================
# LOAD GROUND TRUTH
# ================================================================

print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
)

true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.isna(matched):
        continue

    matched = str(matched).strip()

    if not matched:
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add(
                (s1_id, target_id)
            )

print(
    "Ground-truth positive pairs:",
    f"{len(true_pairs):,}"
)


# ================================================================
# TOKENIZATION
# ================================================================

def tokens(text):

    if not text:
        return []

    return re.findall(
        r"[a-z0-9]+",
        str(text).lower()
    )


def first_token(text):

    t = tokens(text)

    if len(t) == 0:
        return ""

    return t[0]


def last_token(text):

    t = tokens(text)

    if len(t) == 0:
        return ""

    return t[-1]


# ================================================================
# CREATE ALTERNATIVE BLOCKING KEYS
# ================================================================

print("\nCreating alternative blocking keys...")

records = []

for _, row in source1.iterrows():

    s1_id = row["entity_id"]

    country = str(
        row["country"]
    ).strip().lower()

    name_tokens = tokens(
        row["business_name"]
    )

    address_tokens = tokens(
        row["business_address"]
    )

    all_tokens = list(
        dict.fromkeys(
            name_tokens + address_tokens
        )
    )

    keys = set()

    # ------------------------------------------------------------
    # STRATEGY A
    # country + first name token
    # ------------------------------------------------------------
    if country and name_tokens:

        keys.add(
            (
                "A",
                f"{country}|name_first|{name_tokens[0]}"
            )
        )

    # ------------------------------------------------------------
    # STRATEGY B
    # country + last name token
    # ------------------------------------------------------------
    if country and name_tokens:

        keys.add(
            (
                "B",
                f"{country}|name_last|{name_tokens[-1]}"
            )
        )

    # ------------------------------------------------------------
    # STRATEGY C
    # country + first 2 name tokens
    # ------------------------------------------------------------
    if country and len(name_tokens) >= 2:

        keys.add(
            (
                "C",
                f"{country}|name2|"
                f"{name_tokens[0]}|"
                f"{name_tokens[1]}"
            )
        )

    records.append(
        (
            s1_id,
            keys
        )
    )

print(
    "Source-1 records prepared:",
    f"{len(records):,}"
)


# ================================================================
# SQLITE CONNECTION
# ================================================================

print("\nConnecting SQLite...")

conn = sqlite3.connect(
    DB_PATH,
    timeout=60
)

conn.execute(
    "PRAGMA busy_timeout = 60000"
)


# ================================================================
# HELPER
# ================================================================

def generate_candidates(strategy):

    candidate_pairs = set()

    for s1_id, keys in records:

        for key_type, block_key in keys:

            if key_type != strategy:
                continue

            # ----------------------------------------------------
            # IMPORTANT:
            # Existing SQLite index uses normal block_key values
            # such as:
            # country|token
            #
            # For this experiment we cannot directly query
            # name2/name_first/name_last keys from that index.
            #
            # Therefore this strategy is evaluated against the
            # source rows themselves using a temporary in-memory
            # Source-1 grouping, not the production index.
            # ----------------------------------------------------

            pass

    return candidate_pairs


# ================================================================
# IMPORTANT NOTICE
# ================================================================

print("\n" + "=" * 70)
print("CELL 35 PRE-CHECK")
print("=" * 70)

print("""
The existing SQLite blocking index contains keys of the form:

    country|token

Alternative keys such as:

    country|name_first|token
    country|name_last|token
    country|name2|token1|token2

are NOT present in the current database.

Therefore we will NOT run a fake comparison against the
wrong index.

Instead, this cell will only prepare and inspect the
alternative blocking keys safely.
""")

# ------------------------------------------------
# KEY DISTRIBUTION
# ------------------------------------------------

strategy_counts = {
    "A_country_name_first": 0,
    "B_country_name_last": 0,
    "C_country_name2": 0
}

for _, keys in records:

    for key_type, key in keys:

        if key_type == "A":
            strategy_counts[
                "A_country_name_first"
            ] += 1

        elif key_type == "B":
            strategy_counts[
                "B_country_name_last"
            ] += 1

        elif key_type == "C":
            strategy_counts[
                "C_country_name2"
            ] += 1


print("\nAlternative key counts:")

for k, v in strategy_counts.items():

    print(
        f"{k:30s}: {v:,}"
    )


# ================================================================
# SAMPLE KEYS
# ================================================================

print("\n" + "=" * 70)
print("SAMPLE ALTERNATIVE KEYS")
print("=" * 70)

shown = 0

for s1_id, keys in records:

    if not keys:
        continue

    print(
        f"\n{s1_id}"
    )

    for key_type, key in sorted(keys):

        print(
            f"  {key_type}: {key}"
        )

    shown += 1

    if shown >= 10:
        break


# ================================================================
# CLOSE SQLITE
# ================================================================

conn.close()

print("\n" + "=" * 70)
print("CELL 35 COMPLETE")
print("=" * 70)

print("""
No database changes were made.
No large candidate file was generated.
No full dataset was processed.

Next step:
Build these alternative keys into a SMALL TEST INDEX,
then measure recall + candidate count.
""")
```

```python
# ================================================================
# CELL 36 - ALTERNATIVE BLOCKING KEY FEASIBILITY TEST
# ================================================================

import os
import re
import sqlite3
import pandas as pd
from collections import defaultdict

print("=" * 70)
print("CELL 36 - ALTERNATIVE BLOCKING KEY FEASIBILITY TEST")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

SOURCE1_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source1.tsv"
)

GT_PATH = os.path.join(
    BASE_PATH,
    "blocking",
    "candidates",
    "ground_truth_aligned_10k.tsv"
)

DB_PATH = "/content/blocking_lookup.db"

TEST_ROWS = 10_000

# ================================================================
# LOAD SOURCE-1
# ================================================================

print("\nLoading Source-1...")

source1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=TEST_ROWS,
    dtype=str,
    keep_default_na=False
)

print(
    "Source-1 records:",
    f"{len(source1):,}"
)

# ================================================================
# LOAD GROUND TRUTH
# ================================================================

print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
)

true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if pd.isna(matched):
        continue

    matched = str(matched).strip()

    if not matched:
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add(
                (s1_id, target_id)
            )

print(
    "Ground-truth positive pairs:",
    f"{len(true_pairs):,}"
)

# ================================================================
# SQLITE CHECK
# ================================================================

print("\nConnecting SQLite...")

conn = sqlite3.connect(
    DB_PATH,
    timeout=60
)

conn.execute(
    "PRAGMA busy_timeout = 60000"
)

tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name
    """,
    conn
)

print("\nExisting tables:")

print(tables.to_string(index=False))

required = {
    "source2_blocks",
    "source3_blocks"
}

existing = set(
    tables["name"].tolist()
)

if not required.issubset(existing):

    conn.close()

    raise RuntimeError(
        "Required SQLite block tables are missing."
    )

print("\nRequired tables: OK")

# ================================================================
# IMPORTANT:
# EXISTING BLOCK TABLE STRUCTURE
# ================================================================

print("\nChecking block table columns...")

for table in [
    "source2_blocks",
    "source3_blocks"
]:

    info = pd.read_sql_query(
        f"PRAGMA table_info({table})",
        conn
    )

    print(f"\n{table}:")
    print(
        info[
            ["name", "type"]
        ].to_string(index=False)
    )

# ================================================================
# CURRENT BLOCKING KEY LIMITATION
# ================================================================

print("\n" + "=" * 70)
print("KEY STRUCTURE CHECK")
print("=" * 70)

print("""
Current SQLite index stores:

    block_key
    entity_id

The existing block_key is:

    country|token

It does NOT contain:

    business_name
    business_address

Therefore name_first/name_last/name2 cannot be reconstructed
from this index alone.

We will NOT guess or create false mappings.
""")

# ================================================================
# ANALYZE GROUND-TRUTH TARGET IDs
# ================================================================

print("=" * 70)
print("GROUND-TRUTH TARGET ANALYSIS")
print("=" * 70)

target_ids = set()

for _, row in gt.iterrows():

    matched = row["matched_entity_ids"]

    if pd.isna(matched):
        continue

    matched = str(matched).strip()

    if not matched:
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if target_id:
            target_ids.add(target_id)

print(
    "Unique target entity IDs in ground truth:",
    f"{len(target_ids):,}"
)

source2_targets = {
    x for x in target_ids
    if x.startswith("S2-")
}

source3_targets = {
    x for x in target_ids
    if x.startswith("S3-")
}

print(
    "Source-2 target IDs:",
    f"{len(source2_targets):,}"
)

print(
    "Source-3 target IDs:",
    f"{len(source3_targets):,}"
)

# ================================================================
# CHECK TARGET IDS IN CURRENT INDEX
# ================================================================

print("\nChecking target IDs against current indexes...")

def count_existing_targets(table, target_set):

    if not target_set:
        return 0

    found = 0

    # Process in chunks to avoid huge SQL IN clauses
    target_list = list(target_set)

    chunk_size = 500

    for start in range(
        0,
        len(target_list),
        chunk_size
    ):

        chunk = target_list[
            start:start + chunk_size
        ]

        placeholders = ",".join(
            ["?"] * len(chunk)
        )

        query = f"""
        SELECT COUNT(DISTINCT entity_id)
        FROM {table}
        WHERE entity_id IN ({placeholders})
        """

        result = conn.execute(
            query,
            chunk
        ).fetchone()[0]

        found += int(result)

    return found


s2_found = count_existing_targets(
    "source2_blocks",
    source2_targets
)

s3_found = count_existing_targets(
    "source3_blocks",
    source3_targets
)

print(
    "Ground-truth S2 IDs present in index:",
    f"{s2_found:,}"
)

print(
    "Ground-truth S3 IDs present in index:",
    f"{s3_found:,}"
)

# ================================================================
# EXISTING BLOCK COVERAGE
# ================================================================

print("\n" + "=" * 70)
print("CURRENT INDEX COVERAGE")
print("=" * 70)

print("""
The current index can tell us:

    Source-1 blocking key
            ↓
    country|token
            ↓
    S2/S3 entity IDs

But it cannot tell us:

    entity_id
        ↓
    original business_name
        ↓
    first/last/2-token name key

So a real alternative-key test requires the original
Source-2 and Source-3 cleaned TSV files.
""")

# ================================================================
# CLOSE
# ================================================================

conn.close()

print("\n" + "=" * 70)
print("CELL 36 COMPLETE")
print("=" * 70)

print("""
SAFE RESULT:

Existing SQLite index verified.
Ground-truth target coverage checked.
No database changes made.
No candidate explosion.
No large RAM operation.

NEXT:
We need the cleaned Source-2 and Source-3 files to build
a genuine alternative-key index.
""")
```

```python
# ================================================================
# CELL 38 - BUILD ALTERNATIVE KEY TEST INDEX
# ================================================================

import os
import re
import sqlite3
import pandas as pd

print("=" * 70)
print("CELL 38 - BUILDING ALTERNATIVE KEY TEST INDEX")
print("=" * 70)

BASE_PATH = (
    "/content/drive/MyDrive/"
    "6ab10eb3b23ba_student_resource/"
    "student_resource"
)

SOURCE2_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source2.tsv"
)

SOURCE3_PATH = os.path.join(
    BASE_PATH,
    "processed_data",
    "cleaned_train_source3.tsv"
)

ALT_DB = "/content/alternative_blocking_test.db"

# ------------------------------------------------
# REMOVE OLD TEST DATABASE
# ------------------------------------------------

if os.path.exists(ALT_DB):
    os.remove(ALT_DB)

print("\nCreating local SQLite database:")
print(ALT_DB)

conn = sqlite3.connect(
    ALT_DB,
    timeout=120
)

conn.execute("PRAGMA journal_mode=WAL")
conn.execute("PRAGMA synchronous=OFF")
conn.execute("PRAGMA temp_store=FILE")

# ------------------------------------------------
# CREATE TABLE
# ------------------------------------------------

conn.execute("""
CREATE TABLE alternative_blocks (
    block_type TEXT,
    block_key TEXT,
    entity_id TEXT,
    source TEXT
)
""")

print("Table created.")

# ------------------------------------------------
# TOKEN FUNCTION
# ------------------------------------------------

def clean_tokens(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    return [
        x for x in text.split()
        if x
    ]

# ------------------------------------------------
# PROCESS ONE SOURCE
# ------------------------------------------------

def process_source(path, source_name):

    print("\n" + "=" * 60)
    print(f"PROCESSING {source_name}")
    print("=" * 60)

    total = 0
    inserted = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        keep_default_na=False,
        chunksize=50_000
    ):

        # ------------------------------------------------
        # Detect columns
        # ------------------------------------------------

        entity_col = "entity_id"
        name_col = "business_name"

        if entity_col not in chunk.columns:
            raise ValueError(
                f"{entity_col} not found in {source_name}"
            )

        if name_col not in chunk.columns:
            raise ValueError(
                f"{name_col} not found in {source_name}"
            )

        rows = []

        for _, row in chunk.iterrows():

            entity_id = row[entity_col]
            name = row[name_col]

            tokens = clean_tokens(name)

            if not tokens:
                continue

            # --------------------------------------------
            # Alternative A:
            # first name token
            # --------------------------------------------

            first_key = (
                f"name_first|{tokens[0]}"
            )

            rows.append((
                "A",
                first_key,
                entity_id,
                source_name
            ))

            # --------------------------------------------
            # Alternative B:
            # last name token
            # --------------------------------------------

            last_key = (
                f"name_last|{tokens[-1]}"
            )

            rows.append((
                "B",
                last_key,
                entity_id,
                source_name
            ))

            # --------------------------------------------
            # Alternative C:
            # first two tokens
            # --------------------------------------------

            if len(tokens) >= 2:

                two_key = (
                    f"name2|{tokens[0]}|{tokens[1]}"
                )

                rows.append((
                    "C",
                    two_key,
                    entity_id,
                    source_name
                ))

        if rows:

            conn.executemany(
                """
                INSERT INTO alternative_blocks
                (
                    block_type,
                    block_key,
                    entity_id,
                    source
                )
                VALUES (?, ?, ?, ?)
                """,
                rows
            )

            inserted += len(rows)

        total += len(chunk)

        if total % 500_000 == 0:

            conn.commit()

            print(
                f"Processed: {total:,} | "
                f"Index rows: {inserted:,}"
            )

    conn.commit()

    print(
        f"{source_name} complete."
    )

    print(
        f"Records processed: {total:,}"
    )

    print(
        f"Index rows created: {inserted:,}"
    )


# ------------------------------------------------
# BUILD SOURCE-2
# ------------------------------------------------

process_source(
    SOURCE2_PATH,
    "source2"
)

# ------------------------------------------------
# BUILD SOURCE-3
# ------------------------------------------------

process_source(
    SOURCE3_PATH,
    "source3"
)

# ------------------------------------------------
# CREATE INDEX
# ------------------------------------------------

print("\n" + "=" * 70)
print("CREATING SQLITE INDEX")
print("=" * 70)

conn.execute("""
CREATE INDEX idx_alt_block
ON alternative_blocks
(
    block_type,
    block_key
)
""")

conn.execute("""
CREATE INDEX idx_alt_entity
ON alternative_blocks
(
    entity_id
)
""")

conn.commit()

# ------------------------------------------------
# VERIFY
# ------------------------------------------------

print("\n" + "=" * 70)
print("VERIFYING ALTERNATIVE INDEX")
print("=" * 70)

count = conn.execute(
    "SELECT COUNT(*) FROM alternative_blocks"
).fetchone()[0]

print(
    f"Total alternative index rows: {count:,}"
)

summary = pd.read_sql_query(
    """
    SELECT
        source,
        block_type,
        COUNT(*) AS rows
    FROM alternative_blocks
    GROUP BY source, block_type
    ORDER BY source, block_type
    """,
    conn
)

print("\nIndex summary:")
print(summary.to_string(index=False))

conn.close()

size_gb = (
    os.path.getsize(ALT_DB)
    / (1024 ** 3)
)

print("\nDatabase size:")
print(f"{size_gb:.2f} GB")

print("\n" + "=" * 70)
print("CELL 38 COMPLETE")
print("=" * 70)

print("""
Alternative-key index created successfully.

No Source-1 candidate generation was performed.
No full matching was performed.

NEXT:
Test these keys against the 10K Source-1 sample.
""")
```

```python
# ================================================================
# CELL 40 - COUNTRY-AWARE ALTERNATIVE BLOCKING TEST
# ================================================================

import os
import sqlite3
import pandas as pd
import time

print("=" * 70)
print("CELL 40 - COUNTRY-AWARE ALTERNATIVE BLOCKING TEST")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

SOURCE1_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/processed_data/cleaned_train_source1.tsv"
)

GT_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/blocking/candidates/"
    "ground_truth_aligned_10k.tsv"
)

ALT_DB = "/content/alternative_blocking_test.db"

TEST_ROWS = 10_000
MAX_BLOCK_SIZE = 500

# ------------------------------------------------
# CHECK FILES
# ------------------------------------------------

print("\nChecking files...")

for path in [SOURCE1_PATH, GT_PATH, ALT_DB]:
    if not os.path.exists(path):
        raise FileNotFoundError(path)

print("All required files: OK")

# ------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------

print("\nLoading Source-1...")

s1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=TEST_ROWS,
    dtype=str
).fillna("")

print("Source-1 records:", len(s1))

# ------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------

print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
).fillna("")

# Build true pairs
true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if not matched:
        continue

    for target_id in str(matched).split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add((s1_id, target_id))

print("Ground-truth positive pairs:", len(true_pairs))

# ------------------------------------------------
# HELPER: TOKENIZE
# ------------------------------------------------

def normalize_tokens(text):

    text = str(text).lower()

    cleaned = []

    current = ""

    for ch in text:

        if ch.isalnum():
            current += ch
        else:
            if current:
                cleaned.append(current)
                current = ""

    if current:
        cleaned.append(current)

    return cleaned


# ------------------------------------------------
# CREATE COUNTRY + FIRST NAME KEY
# ------------------------------------------------

print("\nCreating country-aware name_first keys...")

source1_keys = []

for _, row in s1.iterrows():

    country = str(row["country"]).strip().lower()

    tokens = normalize_tokens(row["business_name"])

    if tokens:
        first_token = tokens[0]

        block_key = f"{country}|name_first|{first_token}"

        source1_keys.append(
            (
                row["entity_id"],
                block_key
            )
        )

print("Keys created:", len(source1_keys))

# ------------------------------------------------
# CONNECT SQLITE
# ------------------------------------------------

print("\nConnecting alternative SQLite index...")

conn = sqlite3.connect(
    ALT_DB,
    timeout=60
)

print("SQLite connected.")

# ------------------------------------------------
# CHECK TABLE STRUCTURE
# ------------------------------------------------

tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    """,
    conn
)

print("\nTables:")
print(tables)

# Find likely alternative index table
table_names = tables["name"].tolist()

candidate_tables = [
    t for t in table_names
    if "alternative" in t.lower()
    or "block" in t.lower()
]

print("\nCandidate index tables:", candidate_tables)

# ------------------------------------------------
# SELECT INDEX TABLE
# ------------------------------------------------

if not candidate_tables:
    raise RuntimeError(
        "No alternative blocking table found in SQLite database."
    )

ALT_TABLE = candidate_tables[0]

print("\nUsing table:", ALT_TABLE)

# ------------------------------------------------
# SHOW COLUMNS
# ------------------------------------------------

schema = pd.read_sql_query(
    f"PRAGMA table_info({ALT_TABLE})",
    conn
)

print("\nIndex columns:")
print(schema[["name", "type"]])

columns = schema["name"].tolist()

required = ["block_key", "entity_id"]

for col in required:

    if col not in columns:
        raise RuntimeError(
            f"Required column '{col}' not found."
        )

# Detect optional source/block_type columns
has_source = "source" in columns
has_block_type = "block_type" in columns

print("\nSource column:", has_source)
print("Block type column:", has_block_type)

# ------------------------------------------------
# TEST
# ------------------------------------------------

print("\n" + "=" * 70)
print("TESTING COUNTRY-AWARE STRATEGY A")
print("=" * 70)

captured_pairs = set()

candidate_count = 0
records_with_candidates = 0
records_with_capture = 0

start_time = time.time()

for i, (s1_id, block_key) in enumerate(source1_keys, start=1):

    # Query based on available schema
    if has_source and has_block_type:

        sql = f"""
            SELECT entity_id
            FROM {ALT_TABLE}
            WHERE source IN ('source2', 'source3')
              AND block_type = 'A'
              AND block_key = ?
            LIMIT ?
        """

        rows = conn.execute(
            sql,
            (block_key, MAX_BLOCK_SIZE)
        ).fetchall()

    elif has_block_type:

        sql = f"""
            SELECT entity_id
            FROM {ALT_TABLE}
            WHERE block_type = 'A'
              AND block_key = ?
            LIMIT ?
        """

        rows = conn.execute(
            sql,
            (block_key, MAX_BLOCK_SIZE)
        ).fetchall()

    else:

        sql = f"""
            SELECT entity_id
            FROM {ALT_TABLE}
            WHERE block_key = ?
            LIMIT ?
        """

        rows = conn.execute(
            sql,
            (block_key, MAX_BLOCK_SIZE)
        ).fetchall()

    if rows:

        records_with_candidates += 1

        for (candidate_id,) in rows:

            candidate_count += 1

            pair = (
                s1_id,
                candidate_id
            )

            if pair in true_pairs:
                captured_pairs.add(pair)

    if i % 1000 == 0:

        elapsed = (time.time() - start_time) / 60

        print(
            f"Processed: {i:,} | "
            f"Candidates: {candidate_count:,} | "
            f"Captured: {len(captured_pairs):,} | "
            f"Time: {elapsed:.2f} min"
        )

conn.close()

# ------------------------------------------------
# CALCULATE RECALL
# ------------------------------------------------

missed_pairs = len(true_pairs) - len(captured_pairs)

pair_recall = (
    len(captured_pairs) / len(true_pairs) * 100
    if true_pairs
    else 0
)

true_match_records = set(
    s1_id
    for s1_id, _ in true_pairs
)

captured_records = set(
    s1_id
    for s1_id, _ in captured_pairs
)

record_recall = (
    len(captured_records)
    / len(true_match_records)
    * 100
    if true_match_records
    else 0
)

elapsed = (time.time() - start_time) / 60

# ------------------------------------------------
# RESULT
# ------------------------------------------------

print("\n" + "=" * 70)
print("COUNTRY-AWARE STRATEGY A RESULTS")
print("=" * 70)

print(f"Ground-truth positive pairs : {len(true_pairs):,}")
print(f"Captured true pairs         : {len(captured_pairs):,}")
print(f"Missed true pairs           : {missed_pairs:,}")

print(f"\nPair-level recall           : {pair_recall:.2f}%")

print(
    f"\nRecords with >=1 true match : "
    f"{len(true_match_records):,}"
)

print(
    f"Records with >=1 capture    : "
    f"{len(captured_records):,}"
)

print(
    f"Record-level recall         : "
    f"{record_recall:.2f}%"
)

print(f"\nCandidate rows              : {candidate_count:,}")

print(
    f"Average candidates/record   : "
    f"{candidate_count / len(s1):.2f}"
)

print(f"Time taken                  : {elapsed:.2f} minutes")

print("\n" + "=" * 70)
print("CELL 40 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CELL 42 - CORRECT ALTERNATIVE KEY RECALL TEST
# ================================================================

import sqlite3
import pandas as pd
import time

print("=" * 70)
print("CELL 42 - CORRECT ALTERNATIVE KEY RECALL TEST")
print("=" * 70)

SOURCE1_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/processed_data/cleaned_train_source1.tsv"
)

GT_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/blocking/candidates/"
    "ground_truth_aligned_10k.tsv"
)

ALT_DB = "/content/alternative_blocking_test.db"

TEST_ROWS = 10_000
MAX_BLOCK_SIZE = 500

# ------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------

print("\nLoading Source-1...")

s1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    dtype=str,
    nrows=TEST_ROWS
).fillna("")

print("Source-1 records:", len(s1))

# ------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
).fillna("")

true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]
    matched = row["matched_entity_ids"]

    if not matched:
        continue

    for target_id in str(matched).split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add(
                (s1_id, target_id)
            )

print(
    "Ground-truth positive pairs:",
    f"{len(true_pairs):,}"
)

# ------------------------------------------------
# TOKENIZER
# ------------------------------------------------

def normalize_tokens(text):

    text = str(text).lower()

    tokens = []
    current = ""

    for ch in text:

        if ch.isalnum():
            current += ch

        else:

            if current:
                tokens.append(current)
                current = ""

    if current:
        tokens.append(current)

    return tokens


# ------------------------------------------------
# CONNECT DATABASE
# ------------------------------------------------

print("\nConnecting SQLite...")

conn = sqlite3.connect(
    ALT_DB,
    timeout=60
)

conn.execute(
    "PRAGMA busy_timeout=60000"
)

print("SQLite connected.")

# ------------------------------------------------
# TEST STRATEGY A
# ------------------------------------------------

print("\n" + "=" * 70)
print("TESTING A: name_first")
print("=" * 70)

candidate_pairs = set()

candidate_rows = 0
records_with_candidates = 0

start = time.time()

for i, (_, row) in enumerate(
    s1.iterrows(),
    start=1
):

    s1_id = row["entity_id"]

    tokens = normalize_tokens(
        row["business_name"]
    )

    if not tokens:
        continue

    # IMPORTANT:
    # Database actually stores:
    # name_first|TOKEN

    block_key = (
        "name_first|" +
        tokens[0]
    )

    rows = conn.execute(
        """
        SELECT entity_id
        FROM alternative_blocks
        WHERE block_type = 'A'
          AND block_key = ?
        LIMIT ?
        """,
        (
            block_key,
            MAX_BLOCK_SIZE
        )
    ).fetchall()

    if rows:

        records_with_candidates += 1

        for (candidate_id,) in rows:

            candidate_rows += 1

            candidate_pairs.add(
                (
                    s1_id,
                    candidate_id
                )
            )

    if i % 1000 == 0:

        elapsed = (
            time.time() - start
        ) / 60

        print(
            f"Processed: {i:,} | "
            f"Candidates: {candidate_rows:,} | "
            f"Time: {elapsed:.2f} min"
        )

conn.close()

# ------------------------------------------------
# RECALL
# ------------------------------------------------

captured = (
    candidate_pairs &
    true_pairs
)

true_records = {
    x[0]
    for x in true_pairs
}

captured_records = {
    x[0]
    for x in captured
}

pair_recall = (
    len(captured) /
    len(true_pairs) *
    100
)

record_recall = (
    len(captured_records) /
    len(true_records) *
    100
)

avg_candidates = (
    candidate_rows /
    len(s1)
)

elapsed = (
    time.time() - start
) / 60

# ------------------------------------------------
# RESULTS
# ------------------------------------------------

print("\n" + "=" * 70)
print("STRATEGY A RESULTS")
print("=" * 70)

print(
    "Candidate rows:",
    f"{candidate_rows:,}"
)

print(
    "Average candidates/record:",
    f"{avg_candidates:.2f}"
)

print(
    "Captured true pairs:",
    f"{len(captured):,}"
)

print(
    "Missed true pairs:",
    f"{len(true_pairs) - len(captured):,}"
)

print(
    "\nPair-level recall:",
    f"{pair_recall:.2f}%"
)

print(
    "Records with >=1 true match:",
    f"{len(true_records):,}"
)

print(
    "Records with >=1 capture:",
    f"{len(captured_records):,}"
)

print(
    "Record-level recall:",
    f"{record_recall:.2f}%"
)

print(
    "\nTime taken:",
    f"{elapsed:.2f} minutes"
)

print("\n" + "=" * 70)
print("CELL 42 COMPLETE")
print("=" * 70)

print("""
BASELINE:
Pair recall   = 8.23%
Record recall = 10.36%

Compare Strategy A against this baseline.

No database changes were made.
""")
```

```python
# ================================================================
# CELL 46 - OPTIMIZED COMBINED ALTERNATIVE BLOCKING TEST
# name_first + name2
# ================================================================

import os
import sqlite3
import pandas as pd
import re
import time

print("=" * 70)
print("CELL 46 - OPTIMIZED COMBINED BLOCKING TEST")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------
SOURCE1_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/processed_data/cleaned_train_source1.tsv"
)

GT_PATH = (
    "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/"
    "student_resource/blocking/candidates/ground_truth_aligned_10k.tsv"
)

DB_PATH = "/content/alternative_blocking_test.db"

# ------------------------------------------------
# FILE CHECK
# ------------------------------------------------
for path in [SOURCE1_PATH, GT_PATH, DB_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing: {path}")

print("All required files: OK")

# ------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------
print("\nLoading Source-1...")

s1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=10000,
    dtype=str
).fillna("")

print("Source-1 records:", len(s1))

# ------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------
print("\nLoading aligned ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
).fillna("")

print("Ground-truth records:", len(gt))

# ------------------------------------------------
# BUILD TRUE POSITIVE PAIR SET
# ------------------------------------------------
true_pairs = set()
records_with_true_match = set()

for _, row in gt.iterrows():

    s1_id = str(row["source1_entity_id"]).strip()
    matched = str(row["matched_entity_ids"]).strip()

    if not matched or matched.lower() == "nan":
        continue

    ids = [
        x.strip()
        for x in matched.split(",")
        if x.strip()
    ]

    for target_id in ids:
        true_pairs.add((s1_id, target_id))

    if ids:
        records_with_true_match.add(s1_id)

print("Ground-truth positive pairs:", len(true_pairs))
print("Source-1 records with true matches:",
      len(records_with_true_match))

# ------------------------------------------------
# NORMALIZATION FUNCTION
# ------------------------------------------------
def normalize_text(value):
    value = str(value).lower()
    value = re.sub(r"[^a-z0-9]+", " ", value)
    value = re.sub(r"\s+", " ", value).strip()
    return value


# ------------------------------------------------
# CREATE SAME ALTERNATIVE KEYS
# ------------------------------------------------
print("\nCreating Source-1 alternative keys...")

test_keys = []

for _, row in s1.iterrows():

    source1_id = str(row["entity_id"]).strip()

    name = normalize_text(row["business_name"])

    tokens = name.split()

    # -------------------------------
    # Strategy A: name_first
    # -------------------------------
    if len(tokens) >= 1:
        first_token = tokens[0]

        test_keys.append(
            (
                source1_id,
                "A",
                f"name_first|{first_token}"
            )
        )

    # -------------------------------
    # Strategy C: name2
    # -------------------------------
    if len(tokens) >= 2:
        token1 = tokens[0]
        token2 = tokens[1]

        test_keys.append(
            (
                source1_id,
                "C",
                f"name2|{token1}|{token2}"
            )
        )

print("Alternative keys created:", len(test_keys))

# ------------------------------------------------
# CONNECT SQLITE
# ------------------------------------------------
print("\nConnecting alternative SQLite index...")

conn = sqlite3.connect(DB_PATH)

print("SQLite connected.")

# ------------------------------------------------
# CREATE TEMPORARY SOURCE-1 KEY TABLE
# ------------------------------------------------
print("\nCreating temporary key table...")

conn.execute("DROP TABLE IF EXISTS temp_test_keys")

conn.execute("""
CREATE TEMP TABLE temp_test_keys (
    source1_id TEXT,
    block_type TEXT,
    block_key TEXT
)
""")

conn.executemany(
    """
    INSERT INTO temp_test_keys
    (source1_id, block_type, block_key)
    VALUES (?, ?, ?)
    """,
    test_keys
)

conn.execute("""
CREATE INDEX idx_temp_test_keys
ON temp_test_keys(block_type, block_key)
""")

conn.commit()

print("Temporary key index created.")

# ------------------------------------------------
# VERIFY ALTERNATIVE INDEX
# ------------------------------------------------
print("\nChecking alternative index...")

index_count = conn.execute("""
SELECT COUNT(*)
FROM alternative_blocks
""").fetchone()[0]

print("Alternative index rows:", f"{index_count:,}")

# ------------------------------------------------
# OPTIMIZED JOIN
# ------------------------------------------------
print("\nRunning optimized SQLite JOIN...")
print("No large candidate list will be stored in RAM.")

start_time = time.time()

query = """
SELECT DISTINCT
       k.source1_id,
       b.entity_id
FROM temp_test_keys k
JOIN alternative_blocks b
  ON k.block_type = b.block_type
 AND k.block_key = b.block_key
"""

cursor = conn.execute(query)

# ------------------------------------------------
# PROCESS JOIN RESULTS IN CHUNKS
# ------------------------------------------------
captured_pairs = set()
candidate_pair_count = 0

CHUNK_SIZE = 100000

while True:

    rows = cursor.fetchmany(CHUNK_SIZE)

    if not rows:
        break

    candidate_pair_count += len(rows)

    for source1_id, target_id in rows:

        pair = (
            str(source1_id),
            str(target_id)
        )

        if pair in true_pairs:
            captured_pairs.add(pair)

    if candidate_pair_count % 1000000 < CHUNK_SIZE:

        elapsed = (time.time() - start_time) / 60

        print(
            f"Processed candidate pairs: "
            f"{candidate_pair_count:,} | "
            f"Captured true pairs: "
            f"{len(captured_pairs):,} | "
            f"Time: {elapsed:.2f} min"
        )

# ------------------------------------------------
# CALCULATE RECALL
# ------------------------------------------------
missed_pairs = len(true_pairs) - len(captured_pairs)

if len(true_pairs) > 0:
    pair_recall = (
        len(captured_pairs)
        / len(true_pairs)
        * 100
    )
else:
    pair_recall = 0.0

captured_records = {
    pair[0]
    for pair in captured_pairs
}

record_recall = (
    len(captured_records)
    / len(records_with_true_match)
    * 100
    if records_with_true_match
    else 0.0
)

avg_candidates = (
    candidate_pair_count / len(s1)
    if len(s1) > 0
    else 0
)

elapsed = (time.time() - start_time) / 60

# ------------------------------------------------
# FINAL RESULTS
# ------------------------------------------------
print("\n")
print("=" * 70)
print("OPTIMIZED COMBINED RESULTS")
print("=" * 70)

print(f"Test records                  : {len(s1):,}")
print(f"Candidate pairs               : {candidate_pair_count:,}")
print(f"Average candidates/record     : {avg_candidates:.2f}")

print(f"Ground-truth positive pairs   : {len(true_pairs):,}")
print(f"Captured true pairs           : {len(captured_pairs):,}")
print(f"Missed true pairs             : {missed_pairs:,}")

print(f"Pair-level recall             : {pair_recall:.2f}%")

print(f"Records with >=1 true match   : "
      f"{len(records_with_true_match):,}")

print(f"Records with >=1 capture      : "
      f"{len(captured_records):,}")

print(f"Record-level recall           : "
      f"{record_recall:.2f}%")

print(f"Time taken                    : "
      f"{elapsed:.2f} minutes")

print("=" * 70)

# ------------------------------------------------
# COMPARE WITH BASELINES
# ------------------------------------------------
print("\nBASELINE COMPARISON")
print("-" * 70)

print("Original country|token:")
print("  Pair recall   : 8.23%")
print("  Record recall : 10.36%")

print("\nname_first:")
print("  Pair recall   : 38.55%")
print("  Record recall : 54.36%")

print("\nname2:")
print("  Pair recall   : 45.76%")
print("  Record recall : 74.19%")

print("\nCombined name_first + name2:")
print(f"  Pair recall   : {pair_recall:.2f}%")
print(f"  Record recall : {record_recall:.2f}%")

print("=" * 70)

conn.close()

print("\n======================================================================")
print("CELL 46 COMPLETE")
print("======================================================================")
```

```python
# ================================================================
# CELL 48 - FAST COMMON-KEY FILTER TEST
# ================================================================

import sqlite3
import pandas as pd
import time

print("=" * 70)
print("CELL 48 - FAST COMMON-KEY FILTER TEST")
print("=" * 70)

DB_PATH = "/content/alternative_blocking_test.db"

GT_PATH = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource/blocking/candidates/ground_truth_aligned_10k.tsv"

SOURCE1_PATH = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource/processed_data/cleaned_train_source1.tsv"

# ------------------------------------------------
# Load Source-1
# ------------------------------------------------
print("\nLoading Source-1...")

s1 = pd.read_csv(
    SOURCE1_PATH,
    sep="\t",
    nrows=10000,
    dtype=str
).fillna("")

print("Source-1 records:", len(s1))

# ------------------------------------------------
# Load Ground Truth
# ------------------------------------------------
print("\nLoading ground truth...")

gt = pd.read_csv(
    GT_PATH,
    sep="\t",
    dtype=str
).fillna("")

# Build true pair set
true_pairs = set()

for _, row in gt.iterrows():

    s1_id = row["source1_entity_id"]

    matched = str(row["matched_entity_ids"])

    if not matched or matched == "nan":
        continue

    for target_id in matched.split(","):

        target_id = target_id.strip()

        if target_id:
            true_pairs.add((s1_id, target_id))

print("Ground-truth positive pairs:", len(true_pairs))

# ------------------------------------------------
# Connect SQLite
# ------------------------------------------------
print("\nConnecting SQLite...")

conn = sqlite3.connect(DB_PATH)

# ------------------------------------------------
# Check key frequencies
# ------------------------------------------------
print("\nReading key frequencies...")

freq = pd.read_sql_query("""
    SELECT
        block_type,
        source,
        block_key,
        COUNT(*) AS frequency
    FROM alternative_blocks
    GROUP BY block_type, source, block_key
""", conn)

print("Unique keys loaded:", len(freq))

# ------------------------------------------------
# Frequency thresholds
# ------------------------------------------------
thresholds = {
    "A": 1000,   # name_first
    "C": 1000    # name2
}

print("\nFrequency thresholds:")
print("name_first (A) :", thresholds["A"])
print("name2      (C) :", thresholds["C"])

# ------------------------------------------------
# Source-1 key creation
# ------------------------------------------------
def clean_text(x):
    return (
        str(x)
        .lower()
        .strip()
        .replace("|", " ")
    )

def make_tokens(text):
    return [
        x for x in clean_text(text).split()
        if x
    ]

source_keys = {}

for _, row in s1.iterrows():

    sid = row["entity_id"]

    tokens = make_tokens(row["business_name"])

    if not tokens:
        source_keys[sid] = []
        continue

    keys = set()

    # A: first token
    keys.add(("A", "name_first|" + tokens[0]))

    # C: first two tokens
    if len(tokens) >= 2:
        keys.add(("C", "name2|" + tokens[0] + "|" + tokens[1]))

    source_keys[sid] = list(keys)

print("\nSource-1 keys created.")

# ------------------------------------------------
# Build allowed keys
# ------------------------------------------------
allowed_keys = set()

for _, row in freq.iterrows():

    bt = row["block_type"]
    key = row["block_key"]
    frequency = int(row["frequency"])

    if bt in ["A", "C"]:

        if frequency <= thresholds[bt]:
            allowed_keys.add((bt, key))

print("Allowed keys:", len(allowed_keys))

# ------------------------------------------------
# TEST
# ------------------------------------------------
print("\n" + "=" * 70)
print("TESTING FILTERED A + C")
print("=" * 70)

start = time.time()

candidate_pairs = set()
captured = set()

processed = 0

for sid, keys in source_keys.items():

    for bt, key in keys:

        if (bt, key) not in allowed_keys:
            continue

        rows = conn.execute("""
            SELECT entity_id
            FROM alternative_blocks
            WHERE block_type = ?
              AND block_key = ?
        """, (bt, key)).fetchall()

        for (target_id,) in rows:

            pair = (sid, target_id)

            candidate_pairs.add(pair)

            if pair in true_pairs:
                captured.add(pair)

    processed += 1

    if processed % 1000 == 0:
        elapsed = (time.time() - start) / 60

        print(
            f"Processed: {processed:,} | "
            f"Candidates: {len(candidate_pairs):,} | "
            f"Captured: {len(captured):,} | "
            f"Time: {elapsed:.2f} min"
        )

# ------------------------------------------------
# Recall calculation
# ------------------------------------------------
missed = len(true_pairs - captured)

pair_recall = (
    len(captured) / len(true_pairs) * 100
    if true_pairs else 0
)

true_records = set(x[0] for x in true_pairs)
captured_records = set(x[0] for x in captured)

record_recall = (
    len(captured_records) / len(true_records) * 100
    if true_records else 0
)

elapsed = (time.time() - start) / 60

# ------------------------------------------------
# Results
# ------------------------------------------------
print("\n" + "=" * 70)
print("CELL 48 RESULTS")
print("=" * 70)

print(f"Candidate pairs          : {len(candidate_pairs):,}")
print(f"Average candidates/record: {len(candidate_pairs)/10000:.2f}")

print(f"Ground-truth pairs       : {len(true_pairs):,}")
print(f"Captured true pairs      : {len(captured):,}")
print(f"Missed true pairs        : {missed:,}")

print(f"\nPair-level recall        : {pair_recall:.2f}%")

print(f"Records with true match  : {len(true_records):,}")
print(f"Records with capture     : {len(captured_records):,}")

print(f"Record-level recall      : {record_recall:.2f}%")

print(f"\nTime taken               : {elapsed:.2f} minutes")

print("\n" + "=" * 70)
print("BASELINE")
print("=" * 70)

print("Original country|token : Pair 8.23% | Record 10.36%")
print("name_first             : Pair 38.55% | Record 54.36%")
print("name2                  : Pair 45.76% | Record 74.19%")
print("Combined A+C           : Pair 72.79% | Record 94.56%")

print("=" * 70)
print("CELL 48 COMPLETE")
print("=" * 70)

conn.close()
```

```python
# ============================================================
# CELL 49 - FAST FREQUENCY THRESHOLD OPTIMIZATION
# ============================================================

import sqlite3
import pandas as pd
import time
from collections import defaultdict

print("=" * 70)
print("CELL 49 - FAST FREQUENCY THRESHOLD OPTIMIZATION")
print("=" * 70)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------
SOURCE1 = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource/processed_data/cleaned_train_source1.tsv"

GT_PATH = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource/blocking/candidates/ground_truth_aligned_10k.tsv"

DB_PATH = "/content/alternative_blocking_test.db"

# ------------------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------------------
print("\nLoading Source-1...")
s1 = pd.read_csv(SOURCE1, sep="\t", dtype=str).fillna("")
s1 = s1.iloc[:10000].copy()

print("Source-1 records:", len(s1))

# ------------------------------------------------------------
# LOAD GROUND TRUTH
# ------------------------------------------------------------
print("\nLoading ground truth...")
gt = pd.read_csv(GT_PATH, sep="\t", dtype=str).fillna("")

# Build true pairs
true_pairs = set()
true_records = set()

for _, row in gt.iterrows():

    sid = row["source1_entity_id"]
    matched = str(row["matched_entity_ids"]).strip()

    if not matched or matched.lower() == "nan":
        continue

    for tid in matched.split(","):

        tid = tid.strip()

        if not tid:
            continue

        true_pairs.add((sid, tid))
        true_records.add(sid)

print("Ground-truth positive pairs:", len(true_pairs))
print("Source-1 records with true matches:", len(true_records))

# ------------------------------------------------------------
# CREATE SOURCE-1 KEYS
# ------------------------------------------------------------
print("\nCreating Source-1 keys...")

def tokenize(text):
    import re

    text = str(text).lower()
    text = re.sub(r"[^a-z0-9]+", " ", text)

    return [x for x in text.split() if x]


def make_keys(name):

    tokens = tokenize(name)

    if not tokens:
        return []

    # A = first token
    key_a = "name_first|" + tokens[0]

    # C = first two tokens
    if len(tokens) >= 2:
        key_c = "name2|" + tokens[0] + "|" + tokens[1]
    else:
        key_c = "name2|" + tokens[0] + "|"

    return [("A", key_a), ("C", key_c)]


source1_keys = {}

for _, row in s1.iterrows():

    sid = row["entity_id"]

    keys = make_keys(row["business_name"])

    source1_keys[sid] = keys

print("Source-1 key sets created:", len(source1_keys))

# ------------------------------------------------------------
# CONNECT SQLITE
# ------------------------------------------------------------
print("\nConnecting SQLite...")

conn = sqlite3.connect(DB_PATH)

# ------------------------------------------------------------
# THRESHOLDS
# ------------------------------------------------------------
THRESHOLDS = [500, 1000, 2000, 5000, 10000]

results = []

# ------------------------------------------------------------
# TEST EACH THRESHOLD
# ------------------------------------------------------------
for threshold in THRESHOLDS:

    print("\n" + "=" * 60)
    print(f"TESTING FREQUENCY THRESHOLD = {threshold}")
    print("=" * 60)

    start = time.time()

    # --------------------------------------------------------
    # Get allowed keys
    # --------------------------------------------------------
    query = """
        SELECT block_type, block_key
        FROM alternative_blocks
        GROUP BY block_type, block_key
        HAVING COUNT(*) <= ?
    """

    allowed_rows = conn.execute(
        query,
        (threshold,)
    ).fetchall()

    allowed = set(
        (str(a), str(b))
        for a, b in allowed_rows
    )

    print("Allowed keys:", len(allowed))

    # --------------------------------------------------------
    # Candidate + recall calculation
    # --------------------------------------------------------
    candidate_pairs = 0
    captured_pairs = 0

    captured_records = set()

    # Avoid duplicate pair counting
    seen_pairs = set()

    processed = 0

    for sid, keys in source1_keys.items():

        for block_type, block_key in keys:

            if (block_type, block_key) not in allowed:
                continue

            rows = conn.execute(
                """
                SELECT entity_id
                FROM alternative_blocks
                WHERE block_type = ?
                  AND block_key = ?
                """,
                (block_type, block_key)
            ).fetchall()

            for (target_id,) in rows:

                pair = (sid, target_id)

                if pair in seen_pairs:
                    continue

                seen_pairs.add(pair)

                candidate_pairs += 1

                if pair in true_pairs:
                    captured_pairs += 1
                    captured_records.add(sid)

        processed += 1

        if processed % 2000 == 0:
            elapsed = (time.time() - start) / 60

            print(
                f"Processed: {processed:,} | "
                f"Candidates: {candidate_pairs:,} | "
                f"Captured: {captured_pairs:,} | "
                f"Time: {elapsed:.2f} min"
            )

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------
    missed_pairs = len(true_pairs) - captured_pairs

    pair_recall = (
        captured_pairs / len(true_pairs) * 100
        if true_pairs else 0
    )

    record_recall = (
        len(captured_records) / len(true_records) * 100
        if true_records else 0
    )

    avg_candidates = candidate_pairs / len(s1)

    elapsed = (time.time() - start) / 60

    print("\nRESULT")
    print("-" * 60)
    print("Threshold              :", threshold)
    print("Candidate pairs        :", f"{candidate_pairs:,}")
    print("Average candidates/record:", f"{avg_candidates:.2f}")
    print("Captured true pairs    :", f"{captured_pairs:,}")
    print("Missed true pairs      :", f"{missed_pairs:,}")
    print("Pair recall            :", f"{pair_recall:.2f}%")
    print("Record recall          :", f"{record_recall:.2f}%")
    print("Time                   :", f"{elapsed:.2f} min")

    results.append({
        "threshold": threshold,
        "candidate_pairs": candidate_pairs,
        "avg_candidates_per_record": round(avg_candidates, 2),
        "true_pairs": len(true_pairs),
        "captured_pairs": captured_pairs,
        "missed_pairs": missed_pairs,
        "pair_recall_percent": round(pair_recall, 2),
        "record_recall_percent": round(record_recall, 2),
        "time_minutes": round(elapsed, 2)
    })

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------
result_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("CELL 49 - THRESHOLD OPTIMIZATION RESULTS")
print("=" * 70)

print(result_df.to_string(index=False))

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------
OUTPUT = "/content/threshold_optimization_10k.tsv"

result_df.to_csv(
    OUTPUT,
    sep="\t",
    index=False
)

print("\nResults saved to:")
print(OUTPUT)

conn.close()

print("\n" + "=" * 70)
print("CELL 49 COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CELL 50 - FAST BATCH CANDIDATE PREPARATION
# ================================================================

import os
import re
import gc
import sqlite3
import pandas as pd
import time

print("=" * 70)
print("CELL 50 - FAST BATCH CANDIDATE PREPARATION")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

SOURCE1 = f"{BASE}/processed_data/cleaned_train_source1.tsv"
GROUND_TRUTH = f"{BASE}/blocking/candidates/ground_truth_aligned_10k.tsv"
ALT_DB = "/content/alternative_blocking_test.db"

OUTPUT = f"{BASE}/blocking/candidates/ml_candidates_10k.tsv"

# ------------------------------------------------
# SETTINGS
# ------------------------------------------------

MAX_BLOCK_FREQUENCY = 10000

# Maximum candidates retained per Source-1 record.
# This prevents candidate explosion.
MAX_CANDIDATES_PER_RECORD = 500

# ------------------------------------------------
# CHECK FILES
# ------------------------------------------------

for path in [SOURCE1, GROUND_TRUTH, ALT_DB]:

    if not os.path.exists(path):
        raise FileNotFoundError(path)

    print("OK:", path)

# ------------------------------------------------
# LOAD SOURCE-1
# ------------------------------------------------

print("\nLoading Source-1...")

s1 = pd.read_csv(
    SOURCE1,
    sep="\t",
    dtype=str,
    keep_default_na=False
).iloc[:10000].copy()

print("Source-1 records:", len(s1))

# ------------------------------------------------
# IDENTIFY COLUMNS
# ------------------------------------------------

def find_col(df, names):

    lower = {c.lower(): c for c in df.columns}

    for name in names:

        if name.lower() in lower:
            return lower[name.lower()]

    for c in df.columns:

        for name in names:

            if name.lower() in c.lower():
                return c

    return None


ID_COL = find_col(
    s1,
    ["entity_id", "id", "business_id", "record_id"]
)

NAME_COL = find_col(
    s1,
    ["business_name", "name", "company_name"]
)

if ID_COL is None or NAME_COL is None:

    raise ValueError(
        f"Required columns not found.\n"
        f"Columns: {list(s1.columns)}"
    )

print("ID column:", ID_COL)
print("Name column:", NAME_COL)

# ------------------------------------------------
# NORMALIZATION
# ------------------------------------------------

def normalize_text(x):

    x = str(x).lower()

    x = re.sub(
        r"[^a-z0-9]+",
        " ",
        x
    )

    return " ".join(x.split())


def create_keys(name):

    tokens = normalize_text(name).split()

    if len(tokens) == 0:
        return []

    keys = []

    # A: name_first
    keys.append(
        ("A", f"name_first|{tokens[0]}")
    )

    # C: name2
    if len(tokens) >= 2:

        keys.append(
            (
                "C",
                f"name2|{tokens[0]}|{tokens[1]}"
            )
        )

    return keys

# ------------------------------------------------
# CONNECT SQLITE
# ------------------------------------------------

print("\nConnecting SQLite...")

conn = sqlite3.connect(
    ALT_DB
)

# ------------------------------------------------
# CREATE OUTPUT
# ------------------------------------------------

if os.path.exists(OUTPUT):

    os.remove(OUTPUT)

# ------------------------------------------------
# PROCESS RECORD-BY-RECORD
# ------------------------------------------------

print("\nStarting batch candidate preparation...")
print("Maximum block frequency:", MAX_BLOCK_FREQUENCY)
print("Maximum candidates / record:", MAX_CANDIDATES_PER_RECORD)

start = time.time()

output_header = True

total_candidates = 0
processed = 0

batch = []

BATCH_SIZE = 500

# ------------------------------------------------
# MAIN LOOP
# ------------------------------------------------

for _, row in s1.iterrows():

    source1_id = row[ID_COL]

    keys = create_keys(
        row[NAME_COL]
    )

    candidates = {}

    # ------------------------------------------------
    # QUERY A + C
    # ------------------------------------------------

    for block_type, block_key in keys:

        rows = conn.execute(
            """
            SELECT entity_id, source
            FROM alternative_blocks
            WHERE block_type = ?
              AND block_key = ?
            LIMIT ?
            """,
            (
                block_type,
                block_key,
                MAX_CANDIDATES_PER_RECORD
            )
        ).fetchall()

        for target_id, source in rows:

            pair = (
                target_id,
                source
            )

            candidates[pair] = (
                block_type,
                block_key
            )

    # ------------------------------------------------
    # KEEP LIMITED CANDIDATES
    # ------------------------------------------------

    for (target_id, source), (block_type, block_key) in candidates.items():

        batch.append(
            {
                "source1_id": source1_id,
                "target_id": target_id,
                "target_source": source,
                "blocking_type": block_type,
                "blocking_key": block_key
            }
        )

    processed += 1

    # ------------------------------------------------
    # WRITE BATCH
    # ------------------------------------------------

    if processed % BATCH_SIZE == 0:

        if batch:

            batch_df = pd.DataFrame(batch)

            batch_df.drop_duplicates(
                subset=[
                    "source1_id",
                    "target_id"
                ],
                inplace=True
            )

            batch_df.to_csv(
                OUTPUT,
                sep="\t",
                index=False,
                mode="a",
                header=output_header
            )

            output_header = False

            total_candidates += len(batch_df)

            batch.clear()

            del batch_df

            gc.collect()

        elapsed = (
            time.time() - start
        ) / 60

        print(
            f"Processed: {processed:,} | "
            f"Candidates saved: {total_candidates:,} | "
            f"Time: {elapsed:.2f} min"
        )

# ------------------------------------------------
# FINAL BATCH
# ------------------------------------------------

if batch:

    batch_df = pd.DataFrame(batch)

    batch_df.drop_duplicates(
        subset=[
            "source1_id",
            "target_id"
        ],
        inplace=True
    )

    batch_df.to_csv(
        OUTPUT,
        sep="\t",
        index=False,
        mode="a",
        header=output_header
    )

    total_candidates += len(batch_df)

    del batch_df

# ------------------------------------------------
# CLOSE
# ------------------------------------------------

conn.close()

gc.collect()

elapsed = (
    time.time() - start
) / 60

# ------------------------------------------------
# VERIFY
# ------------------------------------------------

print("\n" + "=" * 70)
print("CELL 50 RESULTS")
print("=" * 70)

print(
    "Source-1 records processed:",
    processed
)

print(
    "Candidate rows saved:",
    f"{total_candidates:,}"
)

print(
    "Output file:",
    OUTPUT
)

if os.path.exists(OUTPUT):

    size_gb = (
        os.path.getsize(OUTPUT)
        / (1024 ** 3)
    )

    print(
        "Output size:",
        f"{size_gb:.2f} GB"
    )

print(
    "Time taken:",
    f"{elapsed:.2f} minutes"
)

print("\nSample:")

sample = pd.read_csv(
    OUTPUT,
    sep="\t",
    dtype=str,
    nrows=10
)

print(
    sample.to_string(index=False)
)

print("\n" + "=" * 70)
print("CELL 50 COMPLETE")
print("=" * 70)

print("""
Candidate preparation completed.

Next:
Cell 51 → candidate quality / ML feature preparation.
""")
```

```python
# ======================================================================
# CELL 51R - PERSISTENT ML FEATURE PREPARATION
# ======================================================================

import os
import gc
import time
import pandas as pd

print("=" * 70)
print("CELL 51R - PERSISTENT ML FEATURE PREPARATION")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

CANDIDATE_FILE = f"{BASE}/blocking/candidates/ml_candidates_10k.tsv"
GT_FILE = f"{BASE}/blocking/candidates/ground_truth_aligned_10k.tsv"
OUTPUT_FILE = f"{BASE}/blocking/candidates/ml_features_10k.tsv"

# ------------------------------------------------------------
# CHECK
# ------------------------------------------------------------

for path in [CANDIDATE_FILE, GT_FILE]:
    if not os.path.exists(path):
        raise FileNotFoundError(path)
    print("OK:", path)

# ------------------------------------------------------------
# GROUND TRUTH
# ------------------------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GT_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

print("Ground-truth columns:", list(gt.columns))

positive_pairs = set()

for _, row in gt.iterrows():

    s1 = str(row["source1_entity_id"]).strip()
    targets = str(row["matched_entity_ids"]).strip()

    if not s1 or not targets:
        continue

    targets = (
        targets
        .replace("[", "")
        .replace("]", "")
        .replace("'", "")
        .replace('"', "")
    )

    # Handle multiple target IDs
    for target in targets.replace(";", "|").replace(",", "|").split("|"):

        target = target.strip()

        if target:
            positive_pairs.add((s1, target))

del gt
gc.collect()

print(
    "Ground-truth positive pairs:",
    f"{len(positive_pairs):,}"
)

# ------------------------------------------------------------
# REMOVE OLD INCOMPLETE OUTPUT
# ------------------------------------------------------------

if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

print("\nCreating fresh feature file...")

CHUNK_SIZE = 100_000

total_rows = 0
positive_rows = 0
negative_rows = 0
chunk_no = 0

start = time.time()
first_chunk = True

# ------------------------------------------------------------
# PROCESS CANDIDATES
# ------------------------------------------------------------

for chunk in pd.read_csv(
    CANDIDATE_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False,
    chunksize=CHUNK_SIZE
):

    chunk_no += 1

    # Create pair tuples
    pairs = zip(
        chunk["source1_id"],
        chunk["target_id"]
    )

    chunk["label"] = [
        1 if pair in positive_pairs else 0
        for pair in pairs
    ]

    pos = int((chunk["label"] == 1).sum())
    neg = int((chunk["label"] == 0).sum())

    positive_rows += pos
    negative_rows += neg
    total_rows += len(chunk)

    # Save immediately
    chunk.to_csv(
        OUTPUT_FILE,
        sep="\t",
        index=False,
        mode="w" if first_chunk else "a",
        header=first_chunk
    )

    first_chunk = False

    # Force filesystem visibility
    if chunk_no % 5 == 0:

        size_mb = (
            os.path.getsize(OUTPUT_FILE)
            / (1024 ** 2)
        )

        elapsed = (
            time.time() - start
        ) / 60

        print(
            f"Chunk {chunk_no:02d} | "
            f"Rows: {total_rows:,} | "
            f"Positive: {positive_rows:,} | "
            f"File: {size_mb:.1f} MB | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    gc.collect()

# ------------------------------------------------------------
# FINAL VERIFY
# ------------------------------------------------------------

elapsed = (
    time.time() - start
) / 60

print("\n" + "=" * 70)
print("CELL 51R RESULTS")
print("=" * 70)

print("Feature rows      :", f"{total_rows:,}")
print("Positive rows     :", f"{positive_rows:,}")
print("Negative rows     :", f"{negative_rows:,}")

if os.path.exists(OUTPUT_FILE):

    size_mb = (
        os.path.getsize(OUTPUT_FILE)
        / (1024 ** 2)
    )

    print("Output file       :", OUTPUT_FILE)
    print(f"Output size       : {size_mb:.2f} MB")
    print("FILE SAVED        : ✅ YES")

else:

    raise RuntimeError(
        "Output file was not created!"
    )

print(f"Time taken        : {elapsed:.2f} minutes")

print("=" * 70)
print("CELL 51R COMPLETE")
print("=" * 70)

del positive_pairs
gc.collect()

print("\nMemory cleanup completed.")
```

```python
# ======================================================================
# CELL 52 - RAM-SAFE ML TRAINING DATA PREPARATION
# ======================================================================

import os
import gc
import time
import pandas as pd

print("=" * 70)
print("CELL 52 - RAM-SAFE ML TRAINING DATA PREPARATION")
print("=" * 70)

# ----------------------------------------------------------------------
# PATHS
# ----------------------------------------------------------------------

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

FEATURE_FILE = (
    f"{BASE}/blocking/candidates/ml_features_10k.tsv"
)

TRAIN_FILE = (
    f"{BASE}/blocking/candidates/ml_training_10k.tsv"
)

# ----------------------------------------------------------------------
# CHECK INPUT
# ----------------------------------------------------------------------

print("\nChecking Cell 51R output...")

if not os.path.exists(FEATURE_FILE):
    raise FileNotFoundError(
        f"Cell 51R output not found:\n{FEATURE_FILE}"
    )

feature_size = os.path.getsize(FEATURE_FILE) / (1024 ** 2)

print("FOUND ✅")
print("Feature file:", FEATURE_FILE)
print(f"Feature size: {feature_size:.2f} MB")

# ----------------------------------------------------------------------
# SETTINGS
# ----------------------------------------------------------------------

CHUNK_SIZE = 100_000
NEGATIVE_RATIO = 3

print("\nSettings")
print("-" * 40)
print("Chunk size       :", CHUNK_SIZE)
print("Negative ratio   :", f"{NEGATIVE_RATIO}:1")

# ----------------------------------------------------------------------
# REMOVE OLD TRAINING FILE
# ----------------------------------------------------------------------

if os.path.exists(TRAIN_FILE):
    os.remove(TRAIN_FILE)
    print("\nOld training file removed.")

# ======================================================================
# PASS 1
# COUNT POSITIVE / NEGATIVE ROWS
# ======================================================================

print("\n" + "=" * 70)
print("PASS 1 - COUNTING LABELS")
print("=" * 70)

positive_count = 0
negative_count = 0
total_count = 0

start = time.time()

for chunk_no, chunk in enumerate(
    pd.read_csv(
        FEATURE_FILE,
        sep="\t",
        dtype={
            "source1_id": "string",
            "target_id": "string",
            "target_source": "string",
            "blocking_type": "string",
            "blocking_key": "string",
            "label": "int8"
        },
        chunksize=CHUNK_SIZE,
        keep_default_na=False
    ),
    start=1
):

    positive_count += int(
        (chunk["label"] == 1).sum()
    )

    negative_count += int(
        (chunk["label"] == 0).sum()
    )

    total_count += len(chunk)

    if chunk_no % 10 == 0:

        elapsed = (time.time() - start) / 60

        print(
            f"Processed: {total_count:,} | "
            f"Positive: {positive_count:,} | "
            f"Negative: {negative_count:,} | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    gc.collect()

print("\nPASS 1 COMPLETE")
print("Total rows    :", f"{total_count:,}")
print("Positive rows :", f"{positive_count:,}")
print("Negative rows :", f"{negative_count:,}")

# ----------------------------------------------------------------------
# TARGET NEGATIVE COUNT
# ----------------------------------------------------------------------

target_negative = min(
    negative_count,
    positive_count * NEGATIVE_RATIO
)

print("\nTarget negative samples:", f"{target_negative:,}")

# ======================================================================
# PASS 2
# CREATE BALANCED TRAINING DATA
# ======================================================================

print("\n" + "=" * 70)
print("PASS 2 - CREATING TRAINING DATA")
print("=" * 70)

positive_saved = 0
negative_saved = 0
total_saved = 0

first_write = True

start = time.time()

for chunk_no, chunk in enumerate(
    pd.read_csv(
        FEATURE_FILE,
        sep="\t",
        dtype={
            "source1_id": "string",
            "target_id": "string",
            "target_source": "string",
            "blocking_type": "string",
            "blocking_key": "string",
            "label": "int8"
        },
        chunksize=CHUNK_SIZE,
        keep_default_na=False
    ),
    start=1
):

    # --------------------------------------------------------------
    # ALL POSITIVE ROWS
    # --------------------------------------------------------------

    positive = chunk[
        chunk["label"] == 1
    ]

    # --------------------------------------------------------------
    # NEGATIVE ROWS
    # --------------------------------------------------------------

    negative = chunk[
        chunk["label"] == 0
    ]

    remaining = target_negative - negative_saved

    if remaining <= 0:

        negative = negative.iloc[0:0]

    elif len(negative) > remaining:

        negative = negative.sample(
            n=remaining,
            random_state=42 + chunk_no
        )

    # --------------------------------------------------------------
    # COMBINE
    # --------------------------------------------------------------

    selected = pd.concat(
        [positive, negative],
        ignore_index=True
    )

    # --------------------------------------------------------------
    # SHUFFLE CURRENT CHUNK
    # --------------------------------------------------------------

    if len(selected) > 1:

        selected = selected.sample(
            frac=1,
            random_state=100 + chunk_no
        ).reset_index(drop=True)

    # --------------------------------------------------------------
    # COUNTS
    # --------------------------------------------------------------

    pos_now = int(
        (selected["label"] == 1).sum()
    )

    neg_now = int(
        (selected["label"] == 0).sum()
    )

    positive_saved += pos_now
    negative_saved += neg_now
    total_saved += len(selected)

    # --------------------------------------------------------------
    # SAVE IMMEDIATELY
    # --------------------------------------------------------------

    if len(selected) > 0:

        selected.to_csv(
            TRAIN_FILE,
            sep="\t",
            index=False,
            mode="w" if first_write else "a",
            header=first_write
        )

        first_write = False

    # --------------------------------------------------------------
    # PROGRESS
    # --------------------------------------------------------------

    if chunk_no % 10 == 0:

        elapsed = (time.time() - start) / 60

        print(
            f"Chunk {chunk_no:02d} | "
            f"Saved: {total_saved:,} | "
            f"Positive: {positive_saved:,} | "
            f"Negative: {negative_saved:,} | "
            f"Time: {elapsed:.2f} min"
        )

    # --------------------------------------------------------------
    # MEMORY CLEANUP
    # --------------------------------------------------------------

    del chunk
    del positive
    del negative
    del selected

    gc.collect()

# ======================================================================
# FINAL RESULT
# ======================================================================

elapsed = (time.time() - start) / 60

print("\n" + "=" * 70)
print("CELL 52 RESULTS")
print("=" * 70)

print("Original feature rows :", f"{total_count:,}")
print("Positive samples      :", f"{positive_saved:,}")
print("Negative samples      :", f"{negative_saved:,}")
print("Training rows         :", f"{total_saved:,}")

if positive_saved > 0:

    print(
        "Negative:Positive     :",
        f"{negative_saved / positive_saved:.2f}:1"
    )

if total_saved > 0:

    print(
        "Positive rate         :",
        f"{positive_saved / total_saved * 100:.2f}%"
    )

print("\nOutput file:")
print(TRAIN_FILE)

# ----------------------------------------------------------------------
# VERIFY FILE
# ----------------------------------------------------------------------

if os.path.exists(TRAIN_FILE):

    train_size = (
        os.path.getsize(TRAIN_FILE)
        / (1024 ** 2)
    )

    print(f"\nOutput size           : {train_size:.2f} MB")
    print("FILE SAVED            : ✅ YES")

else:

    raise RuntimeError(
        "Training file was not created!"
    )

print(f"Time taken            : {elapsed:.2f} minutes")

print("=" * 70)
print("CELL 52 COMPLETE")
print("=" * 70)

gc.collect()

print("\nRAM cleanup completed.")
```

```python
# ======================================================================
# CELL 53 - DB-FREE RAM-SAFE ML FEATURE ENGINEERING
# ======================================================================

import os
import gc
import re
import time
import pandas as pd
import numpy as np
from difflib import SequenceMatcher

print("=" * 70)
print("CELL 53 - DB-FREE RAM-SAFE ML FEATURE ENGINEERING")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

SOURCE1_FILE = (
    f"{BASE}/processed_data/cleaned_train_source1.tsv"
)

SOURCE2_FILE = (
    f"{BASE}/processed_data/cleaned_train_source2.tsv"
)

SOURCE3_FILE = (
    f"{BASE}/processed_data/cleaned_train_source3.tsv"
)

CANDIDATE_FILE = (
    f"{BASE}/blocking/candidates/ml_candidates_10k.tsv"
)

GT_FILE = (
    f"{BASE}/blocking/candidates/ground_truth_aligned_10k.tsv"
)

OUTPUT_FILE = (
    f"{BASE}/blocking/candidates/ml_features_10k.tsv"
)

# ----------------------------------------------------------------------
# CHECK FILES
# ----------------------------------------------------------------------

print("\nChecking required files...")

required_files = [
    SOURCE1_FILE,
    SOURCE2_FILE,
    SOURCE3_FILE,
    CANDIDATE_FILE,
    GT_FILE
]

for path in required_files:

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"\nRequired file not found:\n{path}"
        )

    size_mb = os.path.getsize(path) / (1024 ** 2)

    print(
        f"OK: {path} | {size_mb:.2f} MB"
    )

# ----------------------------------------------------------------------
# LOAD SOURCE-1
# ----------------------------------------------------------------------

print("\nLoading Source-1...")

source1 = pd.read_csv(
    SOURCE1_FILE,
    sep="\t",
    dtype=str,
    usecols=lambda c: c in [
        "entity_id",
        "business_name"
    ]
)

if "entity_id" not in source1.columns:
    raise ValueError(
        "Source-1 entity_id column not found"
    )

if "business_name" not in source1.columns:
    raise ValueError(
        "Source-1 business_name column not found"
    )

source1_map = dict(
    zip(
        source1["entity_id"].astype(str),
        source1["business_name"]
        .fillna("")
        .astype(str)
    )
)

print(
    "Source-1 records:",
    f"{len(source1_map):,}"
)

del source1
gc.collect()

# ----------------------------------------------------------------------
# LOAD CANDIDATE TARGET IDs ONLY
# ----------------------------------------------------------------------

print("\nReading candidate IDs...")

candidate_ids = {
    "source2": set(),
    "source3": set()
}

candidate_source1_ids = set()

for chunk in pd.read_csv(
    CANDIDATE_FILE,
    sep="\t",
    dtype=str,
    usecols=[
        "source1_id",
        "target_id",
        "target_source"
    ],
    chunksize=200_000
):

    candidate_source1_ids.update(
        chunk["source1_id"].astype(str)
    )

    for source in ["source2", "source3"]:

        mask = (
            chunk["target_source"].astype(str)
            == source
        )

        if mask.any():

            candidate_ids[source].update(
                chunk.loc[
                    mask,
                    "target_id"
                ].astype(str)
            )

    del chunk
    gc.collect()

print(
    "Unique Source-1 candidate IDs:",
    f"{len(candidate_source1_ids):,}"
)

print(
    "Unique Source-2 target IDs:",
    f"{len(candidate_ids['source2']):,}"
)

print(
    "Unique Source-3 target IDs:",
    f"{len(candidate_ids['source3']):,}"
)

# ----------------------------------------------------------------------
# NORMALIZATION
# ----------------------------------------------------------------------

def normalize_name(x):

    x = str(x).lower()

    x = re.sub(
        r"[^a-z0-9\s]",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    ).strip()

    return x


def ratio(a, b):

    if not a or not b:
        return 0.0

    return (
        SequenceMatcher(
            None,
            a,
            b
        ).ratio()
        * 100
    )


def token_sort_ratio(a, b):

    if not a or not b:
        return 0.0

    a_sorted = " ".join(
        sorted(a.split())
    )

    b_sorted = " ".join(
        sorted(b.split())
    )

    return (
        SequenceMatcher(
            None,
            a_sorted,
            b_sorted
        ).ratio()
        * 100
    )


def token_set_ratio(a, b):

    if not a or not b:
        return 0.0

    a_set = set(a.split())
    b_set = set(b.split())

    if not a_set or not b_set:
        return 0.0

    intersection = a_set & b_set

    if not intersection:
        return 0.0

    common = " ".join(
        sorted(intersection)
    )

    a_full = " ".join(
        sorted(a_set)
    )

    b_full = " ".join(
        sorted(b_set)
    )

    s1 = SequenceMatcher(
        None,
        common,
        a_full
    ).ratio()

    s2 = SequenceMatcher(
        None,
        common,
        b_full
    ).ratio()

    return max(s1, s2) * 100


print("\nSimilarity functions ready ✅")

# ----------------------------------------------------------------------
# LOAD TARGET NAMES CHUNK-WISE
# ----------------------------------------------------------------------

def load_required_targets(
    filepath,
    required_ids,
    source_name
):

    print(
        f"\nScanning {source_name}..."
    )

    result = {}

    # First inspect columns
    header = pd.read_csv(
        filepath,
        sep="\t",
        dtype=str,
        nrows=0
    )

    columns = list(
        header.columns
    )

    print(
        f"{source_name} columns:",
        columns
    )

    id_col = None

    for col in [
        "entity_id",
        "id",
        f"{source_name}_entity_id"
    ]:

        if col in columns:
            id_col = col
            break

    name_col = None

    for col in [
        "business_name",
        "name",
        "name1",
        "name2"
    ]:

        if col in columns:
            name_col = col
            break

    if id_col is None:
        raise ValueError(
            f"ID column not found in {source_name}"
        )

    if name_col is None:
        raise ValueError(
            f"Name column not found in {source_name}"
        )

    print(
        f"{source_name} ID column   : {id_col}"
    )

    print(
        f"{source_name} name column : {name_col}"
    )

    required_ids = set(
        str(x)
        for x in required_ids
    )

    total_scanned = 0

    for chunk in pd.read_csv(
        filepath,
        sep="\t",
        dtype=str,
        usecols=[
            id_col,
            name_col
        ],
        chunksize=100_000
    ):

        chunk[id_col] = (
            chunk[id_col]
            .astype(str)
        )

        mask = chunk[id_col].isin(
            required_ids
        )

        if mask.any():

            selected = chunk.loc[
                mask,
                [id_col, name_col]
            ]

            for row in selected.itertuples(
                index=False,
                name=None
            ):

                result[str(row[0])] = (
                    ""
                    if pd.isna(row[1])
                    else str(row[1])
                )

        total_scanned += len(chunk)

        del chunk
        gc.collect()

    print(
        f"{source_name} scanned:",
        f"{total_scanned:,}"
    )

    print(
        f"{source_name} target names found:",
        f"{len(result):,}"
    )

    return result


source2_map = load_required_targets(
    SOURCE2_FILE,
    candidate_ids["source2"],
    "source2"
)

source3_map = load_required_targets(
    SOURCE3_FILE,
    candidate_ids["source3"],
    "source3"
)

# ----------------------------------------------------------------------
# LOAD GROUND TRUTH
# ----------------------------------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GT_FILE,
    sep="\t",
    dtype=str
)

print(
    "Ground-truth columns:",
    list(gt.columns)
)

if (
    "source1_entity_id" not in gt.columns
    or
    "matched_entity_ids" not in gt.columns
):

    raise ValueError(
        "Unexpected ground-truth format"
    )

positive_pairs = set()

for row in gt.itertuples(
    index=False
):

    source_id = str(
        row.source1_entity_id
    )

    matched_ids = re.split(
        r"[,;|]",
        str(row.matched_entity_ids)
    )

    for target_id in matched_ids:

        target_id = target_id.strip()

        if target_id:

            positive_pairs.add(
                (
                    source_id,
                    target_id
                )
            )

print(
    "Ground-truth positive pairs:",
    f"{len(positive_pairs):,}"
)

del gt
gc.collect()

# ----------------------------------------------------------------------
# REMOVE OLD OUTPUT
# ----------------------------------------------------------------------

if os.path.exists(
    OUTPUT_FILE
):

    os.remove(
        OUTPUT_FILE
    )

# ----------------------------------------------------------------------
# PROCESS CANDIDATES
# ----------------------------------------------------------------------

print("\nCreating ML feature file...")

CHUNK_SIZE = 100_000

total_rows = 0
positive_rows = 0
negative_rows = 0

start_time = time.time()
first_write = True

for chunk_no, chunk in enumerate(
    pd.read_csv(
        CANDIDATE_FILE,
        sep="\t",
        dtype=str,
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    feature_rows = []

    for row in chunk.itertuples(
        index=False
    ):

        source1_id = str(
            row.source1_id
        )

        target_id = str(
            row.target_id
        )

        target_source = str(
            row.target_source
        )

        # --------------------------------------------------------------
        # SOURCE-1 NAME
        # --------------------------------------------------------------

        name1 = source1_map.get(
            source1_id,
            ""
        )

        # --------------------------------------------------------------
        # TARGET NAME
        # --------------------------------------------------------------

        if target_source == "source2":

            name2 = source2_map.get(
                target_id,
                ""
            )

        elif target_source == "source3":

            name2 = source3_map.get(
                target_id,
                ""
            )

        else:

            name2 = ""

        name1 = normalize_name(
            name1
        )

        name2 = normalize_name(
            name2
        )

        # --------------------------------------------------------------
        # NUMERIC FEATURES
        # --------------------------------------------------------------

        name_ratio = ratio(
            name1,
            name2
        )

        token_sort = token_sort_ratio(
            name1,
            name2
        )

        token_set = token_set_ratio(
            name1,
            name2
        )

        length_difference = abs(
            len(name1) - len(name2)
        )

        exact_match = int(
            name1 != ""
            and name2 != ""
            and name1 == name2
        )

        # --------------------------------------------------------------
        # LABEL
        # --------------------------------------------------------------

        label = int(
            (
                source1_id,
                target_id
            )
            in positive_pairs
        )

        if label == 1:

            positive_rows += 1

        else:

            negative_rows += 1

        feature_rows.append({

            "source1_id":
                source1_id,

            "target_id":
                target_id,

            "target_source":
                target_source,

            "blocking_type":
                str(row.blocking_type),

            "name_ratio":
                name_ratio,

            "token_sort_ratio":
                token_sort,

            "token_set_ratio":
                token_set,

            "length_difference":
                length_difference,

            "exact_name_match":
                exact_match,

            "label":
                label
        })

    feature_df = pd.DataFrame(
        feature_rows
    )

    total_rows += len(
        feature_df
    )

    feature_df.to_csv(
        OUTPUT_FILE,
        sep="\t",
        index=False,
        mode="w"
        if first_write
        else "a",
        header=first_write
    )

    first_write = False

    elapsed = (
        time.time() - start_time
    ) / 60

    print(
        f"Chunk {chunk_no:02d} | "
        f"Rows: {total_rows:,} | "
        f"Positive: {positive_rows:,} | "
        f"Time: {elapsed:.2f} min"
    )

    del feature_rows
    del feature_df
    del chunk

    gc.collect()

# ----------------------------------------------------------------------
# FINAL RESULT
# ----------------------------------------------------------------------

elapsed = (
    time.time() - start_time
) / 60

size_mb = (
    os.path.getsize(
        OUTPUT_FILE
    )
    / (1024 ** 2)
)

print("\n" + "=" * 70)
print("CELL 53 RESULTS")
print("=" * 70)

print(
    "Feature rows      :",
    f"{total_rows:,}"
)

print(
    "Positive rows     :",
    f"{positive_rows:,}"
)

print(
    "Negative rows     :",
    f"{negative_rows:,}"
)

print(
    "Positive rate     :",
    f"{positive_rows / total_rows * 100:.4f}%"
)

print("\nNumeric ML features:")
print(" - name_ratio")
print(" - token_sort_ratio")
print(" - token_set_ratio")
print(" - length_difference")
print(" - exact_name_match")

print("\nOutput file:")
print(OUTPUT_FILE)

print(
    f"Output size       : {size_mb:.2f} MB"
)

print(
    "FILE SAVED        : ✅ YES"
)

print(
    f"Time taken        : {elapsed:.2f} minutes"
)

print("=" * 70)
print("CELL 53 COMPLETE")
print("=" * 70)

# ----------------------------------------------------------------------
# CLEANUP
# ----------------------------------------------------------------------

del source1_map
del source2_map
del source3_map
del positive_pairs
gc.collect()

print("\nMemory cleanup completed.")
```

```python
# ======================================================================
# CELL 54R - ROBUST RAM-SAFE TRAIN / VALIDATION PREPARATION
# ======================================================================

import os
import gc
import time
import random
import pandas as pd
import numpy as np

print("=" * 70)
print("CELL 54R - ROBUST RAM-SAFE TRAIN / VALIDATION PREPARATION")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

FEATURE_FILE = (
    f"{BASE}/blocking/candidates/ml_features_10k.tsv"
)

TRAIN_FILE = (
    f"{BASE}/blocking/candidates/ml_train_10k.tsv"
)

VAL_FILE = (
    f"{BASE}/blocking/candidates/ml_validation_10k.tsv"
)

CHUNK_SIZE = 100_000
NEGATIVE_RATIO = 3
VALIDATION_RATIO = 0.20
RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# ----------------------------------------------------------------------
# CHECK FILE
# ----------------------------------------------------------------------

print("\nChecking Cell 53 output...")

if not os.path.exists(FEATURE_FILE):
    raise FileNotFoundError(
        f"Feature file not found:\n{FEATURE_FILE}"
    )

size_mb = os.path.getsize(FEATURE_FILE) / (1024 ** 2)

print("FOUND ✅")
print("Feature file:", FEATURE_FILE)
print(f"Feature size: {size_mb:.2f} MB")

# ----------------------------------------------------------------------
# PASS 1 - COUNT LABELS
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("PASS 1 - COUNTING LABELS")
print("=" * 70)

start = time.time()

total_rows = 0
positive_count = 0
negative_count = 0

# IMPORTANT:
# Do NOT force integer dtype here.
# exact_name_match contains NA values.

for chunk_no, chunk in enumerate(
    pd.read_csv(
        FEATURE_FILE,
        sep="\t",
        chunksize=CHUNK_SIZE,
        low_memory=True
    ),
    start=1
):

    total_rows += len(chunk)

    labels = pd.to_numeric(
        chunk["label"],
        errors="coerce"
    )

    positive_count += int(
        (labels == 1).sum()
    )

    negative_count += int(
        (labels == 0).sum()
    )

    if chunk_no % 10 == 0:
        elapsed = (time.time() - start) / 60

        print(
            f"Processed: {total_rows:,} | "
            f"Positive: {positive_count:,} | "
            f"Negative: {negative_count:,} | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    del labels
    gc.collect()

print("\nPASS 1 COMPLETE")

print(
    "Total rows    :",
    f"{total_rows:,}"
)

print(
    "Positive rows :",
    f"{positive_count:,}"
)

print(
    "Negative rows :",
    f"{negative_count:,}"
)

target_negatives = positive_count * NEGATIVE_RATIO

print(
    "Target negatives:",
    f"{target_negatives:,}"
)

# ----------------------------------------------------------------------
# PASS 2 - COLLECT POSITIVES + SAMPLE NEGATIVES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("PASS 2 - CREATING TRAINING DATA")
print("=" * 70)

start = time.time()

positive_parts = []
negative_reservoir = []

negative_seen = 0

for chunk_no, chunk in enumerate(
    pd.read_csv(
        FEATURE_FILE,
        sep="\t",
        chunksize=CHUNK_SIZE,
        low_memory=True
    ),
    start=1
):

    # --------------------------------------------------------------
    # CLEAN LABEL
    # --------------------------------------------------------------

    chunk["label"] = pd.to_numeric(
        chunk["label"],
        errors="coerce"
    )

    # --------------------------------------------------------------
    # POSITIVES
    # --------------------------------------------------------------

    pos = chunk[
        chunk["label"] == 1
    ]

    if len(pos) > 0:
        positive_parts.append(
            pos.copy()
        )

    # --------------------------------------------------------------
    # NEGATIVES
    # --------------------------------------------------------------

    neg = chunk[
        chunk["label"] == 0
    ]

    for row in neg.itertuples(
        index=False,
        name=None
    ):

        negative_seen += 1

        if len(negative_reservoir) < target_negatives:

            negative_reservoir.append(row)

        else:

            j = random.randrange(
                negative_seen
            )

            if j < target_negatives:
                negative_reservoir[j] = row

    if chunk_no % 10 == 0:

        elapsed = (time.time() - start) / 60

        current_positive = sum(
            len(x)
            for x in positive_parts
        )

        print(
            f"Chunk {chunk_no:02d} | "
            f"Rows: {chunk_no * CHUNK_SIZE:,} approx | "
            f"Positive: {current_positive:,} | "
            f"Negative reservoir: {len(negative_reservoir):,} | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    del pos
    del neg

    gc.collect()

# ----------------------------------------------------------------------
# COMBINE
# ----------------------------------------------------------------------

print("\nCombining samples...")

positive_df = pd.concat(
    positive_parts,
    ignore_index=True
)

del positive_parts
gc.collect()

columns = list(
    positive_df.columns
)

negative_df = pd.DataFrame(
    negative_reservoir,
    columns=columns
)

del negative_reservoir
gc.collect()

print(
    "Positive samples:",
    f"{len(positive_df):,}"
)

print(
    "Negative samples:",
    f"{len(negative_df):,}"
)

# ----------------------------------------------------------------------
# BALANCED DATASET
# ----------------------------------------------------------------------

dataset = pd.concat(
    [
        positive_df,
        negative_df
    ],
    ignore_index=True
)

del positive_df
del negative_df
gc.collect()

# Shuffle
dataset = dataset.sample(
    frac=1,
    random_state=RANDOM_SEED
).reset_index(drop=True)

# ----------------------------------------------------------------------
# TRAIN / VALIDATION SPLIT
# ----------------------------------------------------------------------

validation_size = int(
    len(dataset) * VALIDATION_RATIO
)

validation_df = dataset.iloc[
    :validation_size
].copy()

train_df = dataset.iloc[
    validation_size:
].copy()

del dataset
gc.collect()

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

if os.path.exists(TRAIN_FILE):
    os.remove(TRAIN_FILE)

if os.path.exists(VAL_FILE):
    os.remove(VAL_FILE)

train_df.to_csv(
    TRAIN_FILE,
    sep="\t",
    index=False
)

validation_df.to_csv(
    VAL_FILE,
    sep="\t",
    index=False
)

# ----------------------------------------------------------------------
# VERIFY FILES
# ----------------------------------------------------------------------

train_ok = os.path.exists(TRAIN_FILE)
val_ok = os.path.exists(VAL_FILE)

train_mb = (
    os.path.getsize(TRAIN_FILE) / (1024 ** 2)
    if train_ok else 0
)

val_mb = (
    os.path.getsize(VAL_FILE) / (1024 ** 2)
    if val_ok else 0
)

print("\n" + "=" * 70)
print("CELL 54R RESULTS")
print("=" * 70)

print(
    "Original feature rows :",
    f"{total_rows:,}"
)

print(
    "Positive samples      :",
    f"{positive_count:,}"
)

print(
    "Negative samples      :",
    f"{target_negatives:,}"
)

print(
    "Balanced dataset      :",
    f"{positive_count + target_negatives:,}"
)

print(
    "Negative:Positive     :",
    f"{NEGATIVE_RATIO}:1"
)

print(
    "Training rows         :",
    f"{len(train_df):,}"
)

print(
    "Validation rows       :",
    f"{len(validation_df):,}"
)

print("\nTraining file:")
print(TRAIN_FILE)
print(f"Size: {train_mb:.2f} MB")

print("\nValidation file:")
print(VAL_FILE)
print(f"Size: {val_mb:.2f} MB")

print("\nTraining file saved:", "✅ YES" if train_ok else "❌ NO")
print("Validation file saved:", "✅ YES" if val_ok else "❌ NO")

print("=" * 70)
print("CELL 54R COMPLETE")
print("=" * 70)

del train_df
del validation_df

gc.collect()

print("\nMemory cleanup completed.")
```

```python
# ======================================================================
# CELL 56R - ROBUST RAM-SAFE ML PREDICTION
# ======================================================================

import os
import gc
import time
import joblib
import pandas as pd
import numpy as np

print("=" * 70)
print("CELL 56R - ROBUST RAM-SAFE ML PREDICTION")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

FEATURE_FILE = (
    f"{BASE}/blocking/candidates/ml_features_10k.tsv"
)

MODEL_FILE = (
    f"{BASE}/blocking/candidates/entity_matching_model_10k.pkl"
)

OUTPUT_FILE = (
    f"{BASE}/blocking/candidates/ml_predictions_10k.tsv"
)

CHUNK_SIZE = 100_000
THRESHOLD = 0.50

# ----------------------------------------------------------------------
# CHECK FILES
# ----------------------------------------------------------------------

print("\nChecking required files...")

for path in [FEATURE_FILE, MODEL_FILE]:

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"\nRequired file not found:\n{path}"
        )

    size_mb = os.path.getsize(path) / (1024 ** 2)

    print(
        f"FOUND ✅ {path} | "
        f"{size_mb:.2f} MB"
    )

# ----------------------------------------------------------------------
# LOAD MODEL
# ----------------------------------------------------------------------

print("\nLoading trained model...")

package = joblib.load(MODEL_FILE)

model = package["model"]
FEATURES = package["features"]

print("Model loaded ✅")
print("ML features:", FEATURES)
print("Threshold:", THRESHOLD)

# ----------------------------------------------------------------------
# CHECK FEATURE FILE COLUMNS
# ----------------------------------------------------------------------

print("\nChecking feature columns...")

sample = pd.read_csv(
    FEATURE_FILE,
    sep="\t",
    nrows=5
)

print("Available columns:")
print(list(sample.columns))

missing_features = [
    col for col in FEATURES
    if col not in sample.columns
]

if missing_features:

    raise ValueError(
        "\nRequired ML feature columns missing:\n"
        + str(missing_features)
        + "\nAvailable columns:\n"
        + str(list(sample.columns))
    )

print("All ML feature columns found ✅")

# ----------------------------------------------------------------------
# REMOVE OLD OUTPUT
# ----------------------------------------------------------------------

if os.path.exists(OUTPUT_FILE):

    os.remove(OUTPUT_FILE)

    print(
        "\nOld prediction file removed."
    )

# ----------------------------------------------------------------------
# DETERMINE ID / META COLUMNS
# ----------------------------------------------------------------------

preferred_columns = [
    "source1_id",
    "target_id",
    "target_source",
    "blocking_type",
    "blocking_key",
    "label"
]

available_meta = [
    col
    for col in preferred_columns
    if col in sample.columns
]

print(
    "\nColumns that will be preserved:"
)

print(available_meta)

# ----------------------------------------------------------------------
# PROCESS FILE
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("PREDICTING ALL CANDIDATES")
print("=" * 70)

start_time = time.time()

total_rows = 0
predicted_matches = 0
first_chunk = True

for chunk_no, chunk in enumerate(
    pd.read_csv(
        FEATURE_FILE,
        sep="\t",
        chunksize=CHUNK_SIZE,
        low_memory=True
    ),
    start=1
):

    # --------------------------------------------------------------
    # CLEAN ML FEATURES
    # --------------------------------------------------------------

    for col in FEATURES:

        chunk[col] = pd.to_numeric(
            chunk[col],
            errors="coerce"
        )

    chunk[FEATURES] = (
        chunk[FEATURES]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
    )

    # --------------------------------------------------------------
    # MODEL INPUT
    # --------------------------------------------------------------

    X = chunk[
        FEATURES
    ].astype(
        np.float32
    )

    # --------------------------------------------------------------
    # PREDICTION
    # --------------------------------------------------------------

    probabilities = model.predict_proba(
        X
    )[:, 1]

    predictions = (
        probabilities >= THRESHOLD
    ).astype(
        np.int8
    )

    # --------------------------------------------------------------
    # ADD RESULTS
    # --------------------------------------------------------------

    chunk["match_probability"] = (
        probabilities.astype(
            np.float32
        )
    )

    chunk["prediction"] = predictions

    # --------------------------------------------------------------
    # DYNAMIC OUTPUT COLUMNS
    # --------------------------------------------------------------

    output_columns = [
        col
        for col in available_meta
        if col in chunk.columns
    ]

    output_columns += [
        "match_probability",
        "prediction"
    ]

    output_chunk = chunk[
        output_columns
    ]

    # --------------------------------------------------------------
    # SAVE
    # --------------------------------------------------------------

    output_chunk.to_csv(
        OUTPUT_FILE,
        sep="\t",
        index=False,
        mode="w" if first_chunk else "a",
        header=first_chunk
    )

    first_chunk = False

    # --------------------------------------------------------------
    # COUNTERS
    # --------------------------------------------------------------

    rows = len(chunk)

    total_rows += rows

    predicted_matches += int(
        predictions.sum()
    )

    if chunk_no % 5 == 0:

        elapsed = (
            time.time() - start_time
        ) / 60

        print(
            f"Chunk {chunk_no:02d} | "
            f"Rows: {total_rows:,} | "
            f"Predicted matches: "
            f"{predicted_matches:,} | "
            f"Time: {elapsed:.2f} min"
        )

    # --------------------------------------------------------------
    # MEMORY CLEANUP
    # --------------------------------------------------------------

    del X
    del probabilities
    del predictions
    del output_chunk
    del chunk

    gc.collect()

# ----------------------------------------------------------------------
# FINAL RESULT
# ----------------------------------------------------------------------

elapsed_total = (
    time.time() - start_time
) / 60

output_size = (
    os.path.getsize(OUTPUT_FILE)
    / (1024 ** 2)
)

print("\n" + "=" * 70)
print("CELL 56R RESULTS")
print("=" * 70)

print(
    "Candidate rows processed :",
    f"{total_rows:,}"
)

print(
    "Predicted matches        :",
    f"{predicted_matches:,}"
)

print(
    "Prediction threshold     :",
    THRESHOLD
)

print(
    "Output file:"
)

print(
    OUTPUT_FILE
)

print(
    f"Output size              : "
    f"{output_size:.2f} MB"
)

print(
    f"Time taken               : "
    f"{elapsed_total:.2f} minutes"
)

print(
    "\nFILE SAVED: ✅ YES"
)

print("=" * 70)
print("CELL 56R COMPLETE")
print("=" * 70)

# ----------------------------------------------------------------------
# CLEANUP
# ----------------------------------------------------------------------

del model
del package
del sample

gc.collect()

print("\nMemory cleanup completed.")
```

```python
# ======================================================================
# CELL 58 - THRESHOLD + TOP-K MATCHING COMPARISON
# ======================================================================

import os
import gc
import time
import heapq
import pandas as pd

print("=" * 70)
print("CELL 58 - THRESHOLD + TOP-K MATCHING COMPARISON")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

PREDICTION_FILE = (
    f"{BASE}/blocking/candidates/ml_predictions_10k.tsv"
)

GROUND_TRUTH_FILE = (
    f"{BASE}/blocking/candidates/ground_truth_aligned_10k.tsv"
)

RESULT_FILE = (
    f"{BASE}/blocking/candidates/cell58_threshold_topk_results.tsv"
)

CHUNK_SIZE = 100_000

# We keep only top 5 predictions for each Source-1
TOP_K = 5

THRESHOLDS = [
    0.30,
    0.40,
    0.50,
    0.60,
    0.70,
    0.80,
]

# ----------------------------------------------------------------------
# CHECK FILES
# ----------------------------------------------------------------------

print("\nChecking required files...")

for path in [
    PREDICTION_FILE,
    GROUND_TRUTH_FILE
]:

    if not os.path.exists(path):
        raise FileNotFoundError(path)

    print(
        "FOUND ✅",
        path,
        "|",
        f"{os.path.getsize(path)/(1024**2):.2f} MB"
    )

# ----------------------------------------------------------------------
# LOAD GROUND TRUTH
# ----------------------------------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GROUND_TRUTH_FILE,
    sep="\t",
    dtype=str
)

ground_truth_pairs = set()

for _, row in gt.iterrows():

    source_id = str(
        row["source1_entity_id"]
    ).strip()

    matched_ids = str(
        row["matched_entity_ids"]
    ).strip()

    if not matched_ids:
        continue

    matched_ids = (
        matched_ids
        .replace("|", ",")
        .replace(";", ",")
    )

    for target_id in matched_ids.split(","):

        target_id = target_id.strip()

        if target_id:
            ground_truth_pairs.add(
                (source_id, target_id)
            )

print(
    "Ground-truth pairs:",
    f"{len(ground_truth_pairs):,}"
)

del gt
gc.collect()

# ----------------------------------------------------------------------
# STREAM PREDICTIONS
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("SCANNING PREDICTIONS")
print("=" * 70)

start_time = time.time()

# source1_id -> list of
# (probability, target_id, target_source)
top_matches = {}

total_rows = 0

for chunk_no, chunk in enumerate(
    pd.read_csv(
        PREDICTION_FILE,
        sep="\t",
        chunksize=CHUNK_SIZE,
        low_memory=True
    ),
    start=1
):

    total_rows += len(chunk)

    # Make sure probability is numeric
    chunk["match_probability"] = pd.to_numeric(
        chunk["match_probability"],
        errors="coerce"
    )

    chunk = chunk.dropna(
        subset=["match_probability"]
    )

    for row in chunk.itertuples(
        index=False
    ):

        source_id = str(
            row.source1_id
        )

        target_id = str(
            row.target_id
        )

        target_source = str(
            row.target_source
        )

        probability = float(
            row.match_probability
        )

        if source_id not in top_matches:

            top_matches[source_id] = []

        heap = top_matches[source_id]

        item = (
            probability,
            target_id,
            target_source
        )

        if len(heap) < TOP_K:

            heapq.heappush(
                heap,
                item
            )

        elif probability > heap[0][0]:

            heapq.heapreplace(
                heap,
                item
            )

    if chunk_no % 5 == 0:

        elapsed = (
            time.time() - start_time
        ) / 60

        print(
            f"Chunk {chunk_no:02d} | "
            f"Processed: {total_rows:,} | "
            f"Source-1 tracked: {len(top_matches):,} | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    gc.collect()

print(
    "\nPrediction scan complete."
)

print(
    "Total prediction rows:",
    f"{total_rows:,}"
)

print(
    "Unique Source-1:",
    f"{len(top_matches):,}"
)

# ----------------------------------------------------------------------
# CONVERT HEAPS TO SORTED LISTS
# ----------------------------------------------------------------------

for source_id in top_matches:

    top_matches[source_id].sort(
        reverse=True
    )

# ----------------------------------------------------------------------
# EVALUATE THRESHOLD + TOP-K
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("THRESHOLD + TOP-K RESULTS")
print("=" * 70)

results = []

for threshold in THRESHOLDS:

    for k in [1, 2, 3, 5]:

        selected_pairs = set()

        for source_id, candidates in top_matches.items():

            selected_count = 0

            for probability, target_id, target_source in candidates:

                if probability < threshold:
                    continue

                selected_pairs.add(
                    (
                        source_id,
                        target_id
                    )
                )

                selected_count += 1

                if selected_count >= k:
                    break

        true_positive = len(
            selected_pairs.intersection(
                ground_truth_pairs
            )
        )

        predicted_count = len(
            selected_pairs
        )

        precision = (
            true_positive / predicted_count
            if predicted_count > 0
            else 0
        )

        recall = (
            true_positive / len(
                ground_truth_pairs
            )
            if len(ground_truth_pairs) > 0
            else 0
        )

        f1 = (
            2 * precision * recall /
            (precision + recall)
            if precision + recall > 0
            else 0
        )

        results.append({

            "threshold": threshold,

            "top_k": k,

            "predicted_matches":
                predicted_count,

            "true_positive":
                true_positive,

            "precision":
                precision,

            "recall":
                recall,

            "f1":
                f1
        })

        print(
            f"Threshold={threshold:.2f} | "
            f"Top-K={k} | "
            f"Matches={predicted_count:,} | "
            f"TP={true_positive:,} | "
            f"Precision={precision:.4f} | "
            f"Recall={recall:.4f} | "
            f"F1={f1:.4f}"
        )

# ----------------------------------------------------------------------
# SAVE RESULTS
# ----------------------------------------------------------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "f1",
    ascending=False
)

results_df.to_csv(
    RESULT_FILE,
    sep="\t",
    index=False
)

# ----------------------------------------------------------------------
# BEST CONFIGURATION
# ----------------------------------------------------------------------

best = results_df.iloc[0]

print("\n" + "=" * 70)
print("CELL 58 RESULTS")
print("=" * 70)

print(
    f"Best threshold : "
    f"{best['threshold']:.2f}"
)

print(
    f"Best Top-K     : "
    f"{int(best['top_k'])}"
)

print(
    f"Best Precision  : "
    f"{best['precision']:.4f}"
)

print(
    f"Best Recall     : "
    f"{best['recall']:.4f}"
)

print(
    f"Best F1         : "
    f"{best['f1']:.4f}"
)

print(
    f"Best matches    : "
    f"{int(best['predicted_matches']):,}"
)

print(
    "\nTop configurations:"
)

print(
    results_df.head(5).to_string(
        index=False
    )
)

print(
    "\nOutput file:"
)

print(
    RESULT_FILE
)

print(
    f"Output size: "
    f"{os.path.getsize(RESULT_FILE)/(1024**2):.2f} MB"
)

print("=" * 70)
print("CELL 58 COMPLETE")
print("=" * 70)

# ----------------------------------------------------------------------
# CLEANUP
# ----------------------------------------------------------------------

del top_matches
del ground_truth_pairs
del results
del results_df

gc.collect()

print(
    "\nMemory cleanup completed."
)
```

```python
# ======================================================================
# CELL 59 - FINAL OPTIMIZED MATCHING
# Threshold = 0.80 | Top-K = 5
# ======================================================================

import os
import gc
import heapq
import time
import pandas as pd

print("=" * 70)
print("CELL 59 - FINAL OPTIMIZED MATCHING")
print("=" * 70)

BASE = "/content/drive/MyDrive/6ab10eb3b23ba_student_resource/student_resource"

PREDICTION_FILE = (
    f"{BASE}/blocking/candidates/ml_predictions_10k.tsv"
)

GROUND_TRUTH_FILE = (
    f"{BASE}/blocking/candidates/ground_truth_aligned_10k.tsv"
)

FINAL_FILE = (
    f"{BASE}/blocking/candidates/final_matches_optimized_10k.tsv"
)

THRESHOLD = 0.80
TOP_K = 5
CHUNK_SIZE = 100_000

# ----------------------------------------------------------------------
# CHECK FILES
# ----------------------------------------------------------------------

print("\nChecking required files...")

for path in [
    PREDICTION_FILE,
    GROUND_TRUTH_FILE
]:

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"\nFile not found:\n{path}"
        )

    print(
        "FOUND ✅",
        path,
        "|",
        f"{os.path.getsize(path)/(1024**2):.2f} MB"
    )

# ----------------------------------------------------------------------
# LOAD GROUND TRUTH
# ----------------------------------------------------------------------

print("\nLoading ground truth...")

gt = pd.read_csv(
    GROUND_TRUTH_FILE,
    sep="\t",
    dtype=str
)

ground_truth_pairs = set()

for _, row in gt.iterrows():

    source_id = str(
        row["source1_entity_id"]
    ).strip()

    matched_ids = str(
        row["matched_entity_ids"]
    ).strip()

    if not matched_ids or matched_ids.lower() == "nan":
        continue

    matched_ids = (
        matched_ids
        .replace("|", ",")
        .replace(";", ",")
    )

    for target_id in matched_ids.split(","):

        target_id = target_id.strip()

        if target_id:
            ground_truth_pairs.add(
                (source_id, target_id)
            )

print(
    "Ground-truth pairs:",
    f"{len(ground_truth_pairs):,}"
)

del gt
gc.collect()

# ----------------------------------------------------------------------
# FIND TOP-K PER SOURCE-1
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("SCANNING PREDICTIONS")
print("=" * 70)

start_time = time.time()

top_matches = {}

total_rows = 0
high_confidence_rows = 0

for chunk_no, chunk in enumerate(
    pd.read_csv(
        PREDICTION_FILE,
        sep="\t",
        chunksize=CHUNK_SIZE,
        low_memory=True
    ),
    start=1
):

    total_rows += len(chunk)

    # Safe numeric conversion
    chunk["match_probability"] = pd.to_numeric(
        chunk["match_probability"],
        errors="coerce"
    )

    # Keep only threshold-qualified predictions
    chunk = chunk[
        chunk["match_probability"] >= THRESHOLD
    ]

    high_confidence_rows += len(chunk)

    for row in chunk.itertuples(index=False):

        source_id = str(row.source1_id)
        target_id = str(row.target_id)
        target_source = str(row.target_source)

        probability = float(
            row.match_probability
        )

        if source_id not in top_matches:

            top_matches[source_id] = []

        heap = top_matches[source_id]

        item = (
            probability,
            target_id,
            target_source
        )

        if len(heap) < TOP_K:

            heapq.heappush(
                heap,
                item
            )

        elif probability > heap[0][0]:

            heapq.heapreplace(
                heap,
                item
            )

    if chunk_no % 5 == 0:

        elapsed = (
            time.time() - start_time
        ) / 60

        print(
            f"Chunk {chunk_no:02d} | "
            f"Processed: {total_rows:,} | "
            f"High-confidence: {high_confidence_rows:,} | "
            f"Source-1: {len(top_matches):,} | "
            f"Time: {elapsed:.2f} min"
        )

    del chunk
    gc.collect()

# ----------------------------------------------------------------------
# CREATE FINAL MATCHES
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CREATING FINAL MATCHES")
print("=" * 70)

final_rows = []

for source_id, candidates in top_matches.items():

    candidates.sort(
        reverse=True
    )

    for rank, (
        probability,
        target_id,
        target_source
    ) in enumerate(
        candidates,
        start=1
    ):

        final_rows.append({

            "source1_id":
                source_id,

            "target_id":
                target_id,

            "target_source":
                target_source,

            "match_probability":
                probability,

            "rank":
                rank,

            "threshold":
                THRESHOLD
        })

# ----------------------------------------------------------------------
# SAVE
# ----------------------------------------------------------------------

final_df = pd.DataFrame(
    final_rows
)

final_df.to_csv(
    FINAL_FILE,
    sep="\t",
    index=False
)

# ----------------------------------------------------------------------
# EVALUATION
# ----------------------------------------------------------------------

predicted_pairs = set(
    zip(
        final_df["source1_id"],
        final_df["target_id"]
    )
)

true_positive = len(
    predicted_pairs.intersection(
        ground_truth_pairs
    )
)

predicted_count = len(
    predicted_pairs
)

precision = (
    true_positive / predicted_count
    if predicted_count > 0
    else 0
)

recall = (
    true_positive / len(ground_truth_pairs)
    if ground_truth_pairs
    else 0
)

f1 = (
    2 * precision * recall /
    (precision + recall)
    if precision + recall > 0
    else 0
)

# ----------------------------------------------------------------------
# FINAL RESULTS
# ----------------------------------------------------------------------

elapsed = (
    time.time() - start_time
) / 60

print("\n" + "=" * 70)
print("CELL 59 RESULTS")
print("=" * 70)

print(
    f"Threshold              : {THRESHOLD}"
)

print(
    f"Top-K                  : {TOP_K}"
)

print(
    f"Prediction rows        : {total_rows:,}"
)

print(
    f"High-confidence rows   : {high_confidence_rows:,}"
)

print(
    f"Unique Source-1        : {len(top_matches):,}"
)

print(
    f"Final match rows       : {len(final_df):,}"
)

print(
    f"True positive matches  : {true_positive:,}"
)

print(
    f"Precision              : {precision:.4f}"
)

print(
    f"Recall                 : {recall:.4f}"
)

print(
    f"F1 Score               : {f1:.4f}"
)

print(
    "\nFinal output:"
)

print(
    FINAL_FILE
)

print(
    f"Output size            : "
    f"{os.path.getsize(FINAL_FILE)/(1024**2):.2f} MB"
)

print(
    f"Time taken             : {elapsed:.2f} minutes"
)

print(
    "\nFILE SAVED: ✅ YES"
)

print("=" * 70)
print("CELL 59 COMPLETE")
print("=" * 70)

# ----------------------------------------------------------------------
# CLEANUP
# ----------------------------------------------------------------------

del top_matches
del final_rows
del final_df
del predicted_pairs
del ground_truth_pairs

gc.collect()

print("\nMemory cleanup completed.")
```
